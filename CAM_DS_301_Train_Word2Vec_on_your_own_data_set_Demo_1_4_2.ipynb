## Training your own word embedding from scratch

```python
from datasets import load_dataset
dataset = load_dataset("imdb")
```

```python
dataset
```

```python
text_train = dataset['train']['text']
label_train = dataset['train']['label']
```

```python
text_test = dataset['test']['text']
label_test = dataset['test']['label']
```

```python
from gensim.models import Word2Vec
```

```python
word2vec_texts = text_train
```

```python
word2vec_texts[0]
```

```python
sentences = [sentence.split() for sentence in word2vec_texts]
```

```python
#sentences = [["cat", "say", "meow"], ["dog", "say", "woof"]]
```

```python
# Initialise Word2Vec.
model = Word2Vec(vector_size=100, min_count=1,  window=5)
```

```python
model.build_vocab(sentences)  # Prepare the model vocabulary.
```

```python
model.train(sentences, total_examples=model.corpus_count, epochs=model.epochs)  # Train word vectors.
```

```python
vector = model.wv['king']
```

```python
vector
```

```python
model.wv.most_similar('king', topn=10)  # get other similar words
```

```python
model.wv.most_similar('sweden', topn=10)  # Get other similar words.
```

## Save and use the word embeddings

```python
# Store just the words + their trained embeddings.
word_vectors = model.wv
```

## Use word2vec dense vectors from scratch to train machine learning model

```python
w2v_model = model
import numpy as np
```

```python
# Vectorise the text data.
def vectorize(sentence):
    words = sentence.split()
    words_vecs = [w2v_model.wv[word] for word in words if word in w2v_model.wv]
    if len(words_vecs) == 0:
        return np.zeros(100)
    words_vecs = np.array(words_vecs)
    return words_vecs.mean(axis=0)
```

```python
X_train = np.array([vectorize(sentence) for sentence in text_train ])
```

```python
X_test = np.array([vectorize(sentence) for sentence in text_test])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
# Train a classification model.
clf = LogisticRegression()
clf.fit(X_train, label_train)
```

```python
# Evaluate the model.
y_pred = clf.predict(X_test)
from sklearn.metrics import classification_report

print(classification_report(y_pred, label_test))
```
