# Demonstration 2.1.1 Implementing n-grams

In this demonstration, we will delve into the practical implementation of the n-gram model, an essential concept in natural language processing. We will begin by acquiring the text of *Pride and Prejudice* from Project Gutenberg to serve as our corpus. Using the NLTK library, we will tokenise this text and build models to calculate unigram, bigram, trigram, and four-gram probabilities. We will then explore how to convert these counts into probabilities and demonstrate the difference between deterministic and probabilistic approaches in predicting the next word in a sequence. By the end of this demonstration, we will have a clear understanding of how n-gram models function and their significance in the field of NLP.


```python
import nltk
from nltk.util import ngrams
from collections import defaultdict, Counter
import random
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
import requests
```

```python
# Ensure you have the necessary NLTK data.
nltk.download('punkt')
```

```python
# Download "Pride and Prejudice" from Project Gutenberg.
url = 'https://www.gutenberg.org/files/1342/1342-0.txt'
response = requests.get(url)
corpus = response.text
```

```python
corpus
```

```python
# Tokenise the corpus.
tokens = nltk.word_tokenize(corpus.lower())
```

```python
print(tokens[:50])  # Display the first 50 tokens.
```

```python
# Function to build n-gram model.
def build_ngram_model(tokens, n):
    ngram_counts = defaultdict(Counter)
    for i in range(len(tokens) - n + 1):
        ngram = tuple(tokens[i:i + n - 1])
        next_word = tokens[i + n - 1]
        ngram_counts[ngram][next_word] += 1
    return ngram_counts
```

```python
# Build unigram, bigram, trigram, and four-gram models.
unigram_counts = build_ngram_model(tokens, 1)
bigram_counts = build_ngram_model(tokens, 2)
trigram_counts = build_ngram_model(tokens, 3)
fourgram_counts = build_ngram_model(tokens, 4)
```

```python
bigram_counts
```

```python
trigram_counts
```

```python
# Function to calculate probabilities.
def calculate_probability(ngram_counts):
    probabilities = defaultdict(dict)
    for context, next_words in ngram_counts.items():
        total_count = float(sum(next_words.values()))
        for word, count in next_words.items():
            probabilities[context][word] = count / total_count
    return probabilities
```

```python
# Calculate probabilities for each n-gram model.
unigram_prob = calculate_probability(unigram_counts)
bigram_prob = calculate_probability(bigram_counts)
trigram_prob = calculate_probability(trigram_counts)
fourgram_prob = calculate_probability(fourgram_counts)
```

```python
bigram_prob
```

```python
trigram_prob
```

```python
# Function to predict the next word.
def predict_next_word(sentence, ngram_prob, n, approach = "probabilistic"):
    tokens = nltk.word_tokenize(sentence.lower())
    if n == 1:
        context = tuple()
    else:
        context = tuple(tokens[-(n-1):])
    if approach == 'deterministic':
      if context in ngram_prob:
          next_word = max(ngram_prob[context], key=ngram_prob[context].get)
          return next_word
      else:
          return None
    else:
      if context in ngram_prob:
        next_words = list(ngram_prob[context].keys())
        probabilities = list(ngram_prob[context].values())
        next_word = random.choices(next_words, probabilities)[0]
        return next_word
      else:
        return None
```

```python
input_sentence = "there was reason"
```

```python
# Predict next word using each n-gram model with both deterministic and probabilistic approaches.
print(f'Unigram prediction (deterministic): {predict_next_word(input_sentence, unigram_prob, 1, approach="deterministic")}')
print(f'Bigram prediction (deterministic): {predict_next_word(input_sentence, bigram_prob, 2, approach="deterministic")}')
print(f'Trigram prediction (deterministic): {predict_next_word(input_sentence, trigram_prob, 3, approach="deterministic")}')
print(f'4-gram prediction (deterministic): {predict_next_word(input_sentence, fourgram_prob, 4, approach="deterministic")}')

print(f'Unigram prediction (probabilistic): {predict_next_word(input_sentence, unigram_prob, 1, approach="probabilistic")}')
print(f'Bigram prediction (probabilistic): {predict_next_word(input_sentence, bigram_prob, 2, approach="probabilistic")}')
print(f'Trigram prediction (probabilistic): {predict_next_word(input_sentence, trigram_prob, 3, approach="probabilistic")}')
print(f'4-gram prediction (probabilistic): {predict_next_word(input_sentence, fourgram_prob, 4, approach="probabilistic")}')
```

```python
# Input sentence.
input_sentence = "it is a truth universally"
```
