# Demonstration 2.2.3 Neural networks as universal functional approximators

Follow the demonstration to learn how neural networks can function as universal function approximators. In this demonstration, you'll learn how:
- a neural network with one hidden layer can approximate the sine function
- a neural network with two hidden layers can approximate the sine function

## Neural network with one hidden layer


```python
import numpy as np
import matplotlib.pyplot as plt
from keras.models import Sequential
from keras.layers import Dense
```

```python
# Generate some random data for demonstration.
np.random.seed(42)
X = np.sort(5 * np.random.rand(200, 1), axis=0)
y = np.sin(X).ravel()
```

```python
# Create a neural network with one hidden layer.
model = Sequential()
model.add(Dense(10, input_dim=1, activation='relu'))  # Hidden layer with 10 neurons and ReLU activation
model.add(Dense(1, activation='linear'))  # Output layer with linear activation
```

```python
# Compile the model.
model.compile(loss='mean_squared_error', optimizer='adam')

# Train the model.
model.fit(X, y, epochs=1000, batch_size=20)
```

```python
# Generate predictions using the trained model.
X_test = np.arange(0.0, 5.0, 0.01)[:, np.newaxis]
y_pred = model.predict(X_test)
```

```python
# Check the model shape.
X.shape
```

```python
# Plot the original function and the neural network approximation.
plt.figure(figsize=(8, 6))
plt.scatter(X, y, color='black', label='Original Function')
plt.plot(X_test, y_pred, color='blue', linewidth=2, label='Neural network approximation – Model 1')
plt.title('Neural network as a universal approximator')
plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.show()
```
