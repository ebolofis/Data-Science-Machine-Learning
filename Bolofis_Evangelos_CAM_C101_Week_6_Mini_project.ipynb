```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import datetime
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from typing import List
from scipy.cluster.hierarchy import dendrogram, linkage  # For Hierarchical Clustering
import gc  # Garbage collection

def load_and_explore_data(url: str, chunksize=10**6) -> pd.DataFrame:  # Chunksize added
    """Loads the data from the given URL and performs initial exploration."""

    start_time = datetime.datetime.now()
    print(f"1. load_and_explore_data started at: {start_time}")

    data_chunks = pd.read_csv(url, low_memory=False, chunksize=chunksize)
    combined_data = pd.DataFrame()  # Initialize an empty DataFrame

    for chunk in data_chunks:
        # Data type optimization within the loop
        for col in chunk.columns:
            if pd.api.types.is_numeric_dtype(chunk[col]):
                chunk[col] = pd.to_numeric(chunk[col], downcast='integer')  # Or 'float'
        combined_data = pd.concat([combined_data, chunk], ignore_index=True)  # Concatenate chunks

    print("\nFirst 5 rows:\n", combined_data.head())
    print("\nDataFrame Info:\n", combined_data.info(memory_usage='deep'))  # Detailed memory info
    print("\nDescriptive Statistics:\n", combined_data.describe())

    end_time = datetime.datetime.now()
    print(
        f"1. load_and_explore_data ended at: {end_time}, duration: {end_time - start_time}"
    )
    return combined_data


def clean_currency_values(series: pd.Series) -> pd.Series:
    """Cleans currency values by removing symbols and converting to float."""
    return series.astype(str).str.replace(r"[$,()]", "", regex=True).astype(float)


def cap_outliers_np(
    series: pd.Series, lower_percentile: float = 0.05, upper_percentile: float = 0.95
) -> pd.Series:
    """Caps outliers within the specified percentile range."""
    lower_bound = np.percentile(series, lower_percentile * 100)
    upper_bound = np.percentile(series, upper_percentile * 100)
    return np.clip(series, lower_bound, upper_bound)


def parse_date_robust(series: pd.Series) -> pd.Series:
    """
    Tries multiple date formats to parse a series, returning NaT where parsing fails.
    Handles empty series gracefully.
    """
    if series.empty:
        return pd.Series(dtype='datetime64[ns]')  # Return empty datetime series

    formats = ["%Y-%m-%d", "%d%b%Y", "%m/%d/%Y", "%d/%m/%Y", "%Y/%m/%d", "%b %d, %Y", "%B %d, %Y"]  # Add more formats
    parsed_dates = pd.to_datetime(series, format="ISO8601", errors='coerce')  # Try ISO first

    for fmt in formats:
        parsed_dates = parsed_dates.combine_first(pd.to_datetime(series, format=fmt, errors='coerce'))

    return parsed_dates


def clean_data(data: pd.DataFrame) -> pd.DataFrame:
    """Cleans the data by handling currency, dates, missing values, duplicates, and outliers."""

    start_time = datetime.datetime.now()
    print(f"2. clean_data started at: {start_time}")

    # 2.1 Clean Currency Columns
    currency_cols = ["Total Revenue", "Unit Cost", "Profit"]
    for col in currency_cols:
        data[col] = clean_currency_values(data[col])

    print("\nCleaned Currency Columns:\n", data[currency_cols].head())
    print(data[currency_cols].dtypes)

    # 2.2 Convert Date Columns
    date_cols = ["Order_Date", "Delivery_Date", "Customer_BirthDate"]
    for col in date_cols:
        print(f"\n--- Cleaning Date Column: {col} ---")  # Added: Column header
        print(f"Initial missing values in {col}: {data[col].isnull().sum()}")  # Added: Initial missing value count
        data[col] = parse_date_robust(data[col])
        print(f"Missing values after parsing in {col}: {data[col].isnull().sum()}")  # Added: Missing value count after parsing
        data.dropna(subset=[col], inplace=True)
        print(f"Missing values after dropping NaT in {col}: {data[col].isnull().sum()}")  # Added: Missing value count after dropping NaT
        if data[col].empty:  # Check if column is empty *after* parsing and dropping NaTs
            print(f"WARNING: Column {col} could not be parsed and became empty.")
            data.drop(columns=[col], inplace=True, errors='ignore')  # Remove the column
        if data.empty:
            print("ERROR: DataFrame is empty after date parsing.")
            return pd.DataFrame()

    # 2.3 Handle Missing Values
    print("\n--- Handling Missing Values in Location Data ---")  # Added: Section header
    print(f"Missing values in City before handling: {data['City'].isnull().sum()}")  # Added: Missing value count
    print(f"Missing values in Postal_Code before handling: {data['Postal_Code'].isnull().sum()}")  # Added: Missing value count

    data.dropna(subset=["City", "Postal_Code"], inplace=True)
    print(f"Missing values in City after handling: {data['City'].isnull().sum()}")  # Added: Missing value count after handling
    print(f"Missing values in Postal_Code after handling: {data['Postal_Code'].isnull().sum()}")  # Added: Missing value count after handling

    if data.empty:
        print("ERROR: DataFrame is empty after handling missing values in City/Postal_Code.")
        return pd.DataFrame()

    print(f"\nMissing values in State_Province before handling: {data['State_Province'].isnull().sum()}")  # Added: Missing value count
    data.drop(columns=["State_Province"], inplace=True, errors='ignore')  # Ignore if column doesn't exist
    print("Column 'State_Province' dropped.")  # Added: Explanation
    if data.empty:
        print("ERROR: DataFrame is empty after dropping State_Province.")
        return pd.DataFrame()

    # 2.4 Handle Duplicate Rows
    print("\n--- Handling Duplicate Rows ---")  # Added: Section header
    print(f"Number of duplicate rows before handling: {data.duplicated().sum()}")  # Added: Duplicate count
    data.drop_duplicates(inplace=True)
    print(f"Number of duplicate rows after handling: {data.duplicated().sum()}")  # Added: Duplicate count

    if data.empty:
        print("ERROR: DataFrame is empty after dropping duplicates.")
        return pd.DataFrame()

    # 2.5 Visualize and Cap Outliers
    numerical_cols = data.select_dtypes(include=[np.number]).columns.tolist()
    numerical_cols_to_cap = [
        "Quantity",
        "Total Revenue",
        "Unit Cost",
        "Profit",
        "Days to Delivery",
    ]
    print("\n--- Visualizing and Capping Outliers ---")  # Added: Section header
    print("Boxplots Before Outlier Handling:")  # Added: Explanation
    plt.figure(figsize=(20, 15))
    for i, col in enumerate(numerical_cols):
        plt.subplot(4, 4, i + 1)
        sns.boxplot(y=data[col])
        plt.title(col)
    plt.tight_layout()
    plt.show()

    for col in numerical_cols_to_cap:
        data[col] = cap_outliers_np(data[col])
        print(f"Outliers in '{col}' capped at 5th and 95th percentiles.")  # Added: Explanation

    print("\nBoxplots After Outlier Handling:")  # Added: Explanation
    plt.figure(figsize=(20, 15))
    for i, col in enumerate(numerical_cols_to_cap):
        plt.subplot(3, 2, i + 1)
        sns.boxplot(y=data[col])
        plt.title(col)
    plt.tight_layout()
    plt.show()

    end_time = datetime.datetime.now()
    print(f"2. clean_data ended at: {end_time}, duration: {end_time - start_time}")
    return data


# 3. Feature Engineering

def create_features(data: pd.DataFrame) -> pd.DataFrame:
    """Creates new features for customer segmentation."""

    start_time = datetime.datetime.now()
    print(f"3. create_features started at: {start_time}")

    agg_data = data.groupby("Customer ID")

    frequency_df = agg_data.size().reset_index(name="Frequency")
    print("\nCalculated Frequency.")

    latest_date = data["Order_Date"].max()
    recency_df = agg_data["Order_Date"].max().reset_index()
    if 'Order_Date' in data:
        recency_df["Recency"] = (latest_date - recency_df["Order_Date"]).dt.days
        print("Calculated Recency.")
    else:
        recency_df["Recency"] = 0
        print("WARNING: Order_Date column missing. Recency set to 0.")
    recency_df.drop(columns=["Order_Date"], inplace=True, errors='ignore')

    revenue_per_customer = agg_data["Total Revenue"].sum().reset_index()
    order_counts_per_customer = agg_data.size().reset_index(name="OrderCount")
    clv_df = revenue_per_customer.merge(order_counts_per_customer, on="Customer ID")
    clv_df["AvgPurchaseValue"] = clv_df["Total Revenue"] / clv_df["OrderCount"]
    clv_df["CLV"] = clv_df["AvgPurchaseValue"] * clv_df["OrderCount"] / (3 * 365)
    clv_df.drop(columns=["AvgPurchaseValue", "OrderCount"], inplace=True, errors='ignore')
    print("Calculated CLV.")

    avg_unit_cost_df = agg_data.apply(
        lambda x: x["Total Revenue"].sum() / x["Quantity"].sum()
    ).reset_index(name="AvgUnitCost")
    print("Calculated Average Unit Cost.")

    age_df = data[["Customer ID", "Customer_BirthDate"]].drop_duplicates()
    if 'Order_Date' in data and 'Customer_BirthDate' in data:
        latest_date = data["Order_Date"].max()
        age_df["Customer_Age"] = (
            (latest_date - age_df["Customer_BirthDate"]).dt.days
        ) / 365
        print("Calculated Customer Age.")
    else:
        age_df["Customer_Age"] = 0
        print("WARNING: Order_Date or Customer_BirthDate column(s) missing. Customer Age set to 0.")
    age_df.drop(columns=["Customer_BirthDate"], inplace=True, errors='ignore')

    feature_df = frequency_df.merge(recency_df, on="Customer ID", how="left")
    feature_df = feature_df.merge(clv_df, on="Customer ID", how="left")
    feature_df = feature_df.merge(avg_unit_cost_df, on="Customer ID", how="left")
    feature_df = feature_df.merge(age_df, on="Customer ID", how="left")
    print("\nMerged all features into feature_df.")

    print("\nFeature DataFrame:\n", feature_df.head())
    print("\nFeature DataFrame Info:\n", feature_df.info(memory_usage='deep'))  # Detailed memory info

    end_time = datetime.datetime.now()
    print(f"3. create_features ended at: {end_time}, duration: {end_time - start_time}")

    # Explicitly delete intermediate DataFrames to free memory
    del frequency_df, recency_df, clv_df, avg_unit_cost_df, age_df
    gc.collect()  # Run garbage collector
    return feature_df

# 4. EDA of Engineered Features
def explore_engineered_features(feature_df: pd.DataFrame):
    """Performs EDA on the engineered features."""

    start_time = datetime.datetime.now()
    print(f"4. explore_engineered_features started at: {start_time}")

    print("\n--- 4. EDA of Engineered Features ---")
    print("\nDescriptive Statistics of Engineered Features:\n", feature_df.describe())
    feature_df.hist(figsize=(12, 10))
    plt.suptitle("Distribution of Engineered Features", y=1.02)
    plt.tight_layout()
    plt.show()

    # Add boxplots of engineered features
    plt.figure(figsize=(20, 15))
    features_to_plot = ["Frequency", "Recency", "CLV", "AvgUnitCost", "Customer_Age"]
    for i, col in enumerate(features_to_plot):
        plt.subplot(3, 2, i + 1)
        sns.boxplot(y=feature_df[col])
        plt.title(col)
    plt.tight_layout()
    plt.show()

    end_time = datetime.datetime.now()
    print(
        f"4. explore_engineered_features ended at: {end_time}, duration: {end_time - start_time}"
    )


# 5. Feature Scaling
def scale_features(feature_df: pd.DataFrame, features_to_scale: List[str]) -> pd.DataFrame:
    """Scales the specified features using StandardScaler."""

    start_time = datetime.datetime.now()
    print(f"5. scale_features started at: {start_time}")

    scaler = StandardScaler()
    feature_df[features_to_scale] = scaler.fit_transform(feature_df[features_to_scale])

    end_time = datetime.datetime.now()
    print(f"5. scale_features ended at: {end_time}, duration: {end_time - start_time}")
    return feature_df


# 6. Determine Optimal Number of Clusters (k)
def find_optimal_k(feature_df: pd.DataFrame, features_to_scale: List[str], k_values: List[int]):
    """Finds the optimal number of clusters (k) using the Elbow method and Silhouette analysis."""

    start_time = datetime.datetime.now()
    print(f"6. find_optimal_k started at: {start_time}")

    wcss = []
    silhouette_scores = []

    for k in k_values:
        kmeans = KMeans(
            n_clusters=k, random_state=42, n_init="auto"
        )
        kmeans.fit(feature_df[features_to_scale])
        wcss.append(kmeans.inertia_)
        if k > 2:
            silhouette_scores.append(
                silhouette_score(feature_df[features_to_scale], kmeans.labels_)
            )

    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.plot(k_values, wcss, marker="o")
    plt.title("Elbow Method for Optimal k")
    plt.xlabel("Number of Clusters (k)")
    plt.ylabel("WCSS")

    plt.subplot(1, 2, 2)
    plt.plot(
        k_values[1:], silhouette_scores, marker="o"
    )
    plt.title("Silhouette Score for Optimal k")
    plt.xlabel("Number of Clusters (k)")
    plt.ylabel("Silhouette Score")

    plt.tight_layout()
    plt.show()

    print("\nSilhouette Scores:", silhouette_scores)

    # Hierarchical Clustering and Dendrogram
    linked = linkage(feature_df[features_to_scale], method='ward')
    plt.figure(figsize=(12, 6))
    dendrogram(linked, orientation='top', labels=feature_df['Customer ID'].tolist() if 'Customer ID' in feature_df else None)  # adding labels
    plt.title('Dendrogram for Hierarchical Clustering')
    plt.xlabel('Customer ID')
    plt.ylabel('Cluster Distance')
    plt.show()

    end_time = datetime.datetime.now()
    print(f"6. find_optimal_k ended at: {end_time}, duration: {end_time - start_time}")


# 7. Perform K-Means Clustering and Analyze Clusters
def perform_kmeans_clustering(
    feature_df: pd.DataFrame, features_to_scale: List[str], optimal_k: int
) -> pd.DataFrame:
    """Performs K-means clustering and analyzes the resulting clusters."""

    start_time = datetime.datetime.now()
    print(f"7. perform_kmeans_clustering started at: {start_time}")

    kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init="auto")
    feature_df["Cluster"] = kmeans.fit_predict(feature_df[features_to_scale])

    cluster_assignment = feature_df[['Customer ID', 'Cluster']]  # Create cluster assignment table
    print("\nCluster Assignments:\n", cluster_assignment.head(10))  # Show the first 10 assignments

    end_time = datetime.datetime.now()
    print(
        f"7. perform_kmeans_clustering ended at: {end_time}, duration: {end_time - start_time}"
    )
    return feature_df

def visualize_cluster_analysis(

        feature_df: pd.DataFrame, features_to_visualize: List[str]
):
    """Visualizes the cluster analysis using boxplots."""

    start_time = datetime.datetime.now()
    print(f"8. visualize_cluster_analysis started at: {start_time}")

    fig, axes = plt.subplots(
        1, len(features_to_visualize), figsize=(25, 5), sharey=False
    )
    fig.suptitle("Cluster Analysis: Feature Distributions", fontsize=16)

    for i, feature in enumerate(features_to_visualize):
        sns.boxplot(ax=axes[i], x="Cluster", y=feature, data=feature_df)
        axes[i].set_title(feature, fontsize=12)
        axes[i].set_xlabel("Cluster", fontsize=10)
        axes[i].set_ylabel("Value", fontsize=10)

    plt.tight_layout()
    plt.show()

    end_time = datetime.datetime.now()
    print(
        f"8. visualize_cluster_analysis ended at: {end_time}, duration: {end_time - start_time}"
    )


# 8. Dimensionality Reduction and Visualization
def perform_dimensionality_reduction(
    feature_df: pd.DataFrame, features_to_scale: List[str]
) -> pd.DataFrame:
    """Performs dimensionality reduction using PCA and t-SNE."""

    start_time = datetime.datetime.now()
    print(f"9. perform_dimensionality_reduction started at: {start_time}")

    pca = PCA(n_components=2, random_state=42)
    feature_df["PCA1"], feature_df["PCA2"] = pca.fit_transform(
        feature_df[features_to_scale]
    ).T

    tsne = TSNE(
        n_components=2,
        n_iter=250,
        n_jobs=-1,
        init='pca',
        random_state=42
    )
    feature_df["TSNE1"], feature_df["TSNE2"] = tsne.fit_transform(
        feature_df[features_to_scale]
    ).T

    end_time = datetime.datetime.now()
    print(
        f"9. perform_dimensionality_reduction ended at: {end_time}, duration: {end_time - start_time}"
    )
    return feature_df


def visualize_clusters_2d(feature_df: pd.DataFrame):
    """Visualizes clusters in 2D using PCA and t-SNE results."""

    start_time = datetime.datetime.now()
    print(f"10. visualize_clusters_2d started at: {start_time}")

    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    sns.scatterplot(x="PCA1", y="PCA2", hue="Cluster", data=feature_df, palette="viridis")
    plt.title("PCA Visualization of Clusters")
    plt.xlabel("PCA Component 1")
    plt.ylabel("PCA Component 2")

    plt.subplot(1, 2, 2)
    sns.scatterplot(
        x="TSNE1", y="TSNE2", hue="Cluster", data=feature_df, palette="viridis"
    )
    plt.title("t-SNE Visualization of Clusters")
    plt.xlabel("t-SNE Component 1")
    plt.ylabel("t-SNE Component 2")

    plt.tight_layout()
    plt.show()

    end_time = datetime.datetime.now()
    print(
        f"10. visualize_clusters_2d ended at: {end_time}, duration: {end_time - start_time}"
    )


# Main Execution
def main():
    start_time_main = datetime.datetime.now()
    print(f"Main Execution started at: {start_time_main}")

    data_url = "https://github.com/FourthRevGitHub/CAM_DS_Public-Data-Sets/raw/main/Course%201/Week%206/CUSTOMERS_CLEAN.zip"
    data = load_and_explore_data(data_url)
    if data.empty:
        print("Error: Data loading failed. Exiting.")
        return
    data = clean_data(data)
    if data.empty:
        print("Error: Data cleaning failed. Exiting.")
        return
    feature_df = create_features(data)
    explore_engineered_features(feature_df)
    features_to_scale = ["Frequency", "Recency", "CLV", "AvgUnitCost", "Customer_Age"]
    feature_df = scale_features(feature_df, features_to_scale)
    k_values = range(2, 11)
    find_optimal_k(feature_df, features_to_scale, k_values)
    optimal_k = 4
    feature_df = perform_kmeans_clustering(feature_df, features_to_scale, optimal_k)
    features_to_visualize = ["Frequency", "Recency", "CLV", "AvgUnitCost", "Customer_Age"]
    visualize_cluster_analysis(feature_df, features_to_visualize)
    feature_df = perform_dimensionality_reduction(feature_df, features_to_scale)
    visualize_clusters_2d(feature_df)
    # 9. Save the final DataFrame
    feature_df.to_csv("customer_segmentation_data.csv", index=False)
    print("\nFinal DataFrame saved to 'customer_segmentation_data.csv'")
    end_time_main = datetime.datetime.now()
    print(f"Main Execution ended at: {end_time_main}, duration: {end_time_main - start_time_main}")


if __name__ == "__main__":
    main()
```
