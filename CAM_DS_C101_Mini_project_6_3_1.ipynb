```python
import pandas as pd
import numpy as np
import datetime
from typing import Iterator, Optional, Tuple, List
import gc
from sklearn.preprocessing import StandardScaler

def load_and_explore_data(url: str, chunksize=10**4) -> Iterator[pd.DataFrame]:
    """Loads and explores data in chunks, yielding cleaned chunks."""

    start_time = datetime.datetime.now()
    print(f"1. load_and_explore_data started: {start_time}")

    data_chunks = pd.read_csv(url, low_memory=False, chunksize=chunksize)

    for i, chunk in enumerate(data_chunks):
        print(f"\n--- load_and_explore_data: Processing Chunk {i + 1} ---")

        # 1.  Data type optimization
        for col in chunk.columns:
            if pd.api.types.is_numeric_dtype(chunk[col]):
                chunk[col] = pd.to_numeric(chunk[col], downcast="integer")
            if col in ["Order_Date", "Delivery_Date", "Customer_BirthDate"]:
                chunk[col] = pd.to_datetime(chunk[col], errors='coerce')

        if i == 0:  # Basic exploration (first chunk)
            print("\nFirst 5 rows:\n", chunk.head())
            print("\nDataFrame Info:\n", chunk.info(memory_usage="deep"))
            print("\nDescriptive Statistics:\n", chunk.describe())

        yield chunk
        del chunk
        gc.collect()

    end_time = datetime.datetime.now()
    print(f"1. load_and_explore_data ended: {end_time}, duration: {end_time - start_time}")

def clean_currency_values(series: pd.Series) -> pd.Series:
    """Cleans currency values by removing symbols and converting to float."""

    return series.astype(str).str.replace(r"[$,()]", "", regex=True).astype(float)

def cap_outliers_np(series: pd.Series, lower_percentile: float = 0.05, upper_percentile: float = 0.95) -> pd.Series:
    """Caps outliers within the specified percentile range."""

    lower_bound = np.percentile(series, lower_percentile * 100)
    upper_bound = np.percentile(series, upper_percentile * 100)
    return np.clip(series, lower_bound, upper_bound)

def clean_data(data_iterator: Iterator[pd.DataFrame]) -> Iterator[pd.DataFrame]:
    """Cleans data in chunks."""

    start_time = datetime.datetime.now()
    print(f"2. clean_data started at: {start_time}")

    for i, chunk in enumerate(data_iterator):
        print(f"\n--- clean_data: Processing Chunk {i + 1} ---")

        # 2.  Currency columns
        currency_cols = ["Total Revenue", "Unit Cost", "Profit"]
        for col in currency_cols:
            if col in chunk:
                chunk[col] = clean_currency_values(chunk[col])
                if pd.api.types.is_numeric_dtype(chunk[col]):
                    chunk[col] = pd.to_numeric(chunk[col], downcast="float")

        # 3.  Date columns (NaTs handled in load_and_explore)
        date_cols = ["Order_Date", "Delivery_Date", "Customer_BirthDate"]
        for col in date_cols:
            if col in chunk:
                chunk.dropna(subset=[col], inplace=True)

        # 4.  Missing values
        chunk.dropna(subset=["City", "Postal_Code"], inplace=True)
        if "State_Province" in chunk:
            chunk.drop(columns=["State_Province"], inplace=True, errors="ignore")

        # 5.  Duplicates
        chunk.drop_duplicates(inplace=True)

        # 6.  Outliers
        numerical_cols_to_cap = ["Quantity", "Total Revenue", "Unit Cost", "Profit", "Days to Delivery"]
        for col in numerical_cols_to_cap:
            if col in chunk:
                chunk[col] = cap_outliers_np(chunk[col])

        yield chunk
        del chunk
        gc.collect()

    end_time = datetime.datetime.now()
    print(f"2. clean_data ended at: {end_time}, duration: {end_time - start_time}")

def calculate_frequency(data_iterator: Iterator[pd.DataFrame]) -> Iterator[pd.DataFrame]:
    """Calculates frequency (order counts) and yields chunks."""

    for i, chunk in enumerate(data_iterator):
        print(f"  - calculate_frequency: Processing Chunk {i + 1}")
        if 'Customer ID' in chunk:
          frequency_chunk = chunk.groupby("Customer ID").size().reset_index(name="Frequency")
          yield frequency_chunk
          del frequency_chunk, chunk
          gc.collect()
        else:
          print("WARNING: Customer ID column missing in chunk")
          yield pd.DataFrame()
          del chunk
          gc.collect()

def calculate_recency(data_iterator: Iterator[pd.DataFrame]) -> Iterator[pd.DataFrame]:
    """Calculates recency (days since last order) and yields chunks."""

    for i, chunk in enumerate(data_iterator):
        print(f"  - calculate_recency: Processing Chunk {i + 1}")
        if 'Customer ID' in chunk and 'Order_Date' in chunk and not chunk['Order_Date'].empty:
          recency_chunk = chunk.groupby("Customer ID")["Order_Date"].max().reset_index()
          recency_latest_date = chunk["Order_Date"].max()
          recency_chunk['Latest_Date'] = recency_latest_date
          yield recency_chunk
          del recency_chunk, recency_latest_date, chunk
          gc.collect()
        else:
          print("WARNING: Customer ID or Order_Date column missing or empty in chunk")
          yield pd.DataFrame()
          del chunk
          gc.collect()

def calculate_clv(data_iterator: Iterator[pd.DataFrame]) -> Iterator[pd.DataFrame]:
    """Calculates CLV (Customer Lifetime Value) and yields chunks."""

    for i, chunk in enumerate(data_iterator):
        print(f"  - calculate_clv: Processing Chunk {i + 1}")
        if 'Customer ID' in chunk and 'Total Revenue' in chunk:
          revenue_chunk = chunk.groupby("Customer ID")["Total Revenue"].sum().reset_index(name="TotalRevenue")
          order_count_chunk = chunk.groupby("Customer ID").size().reset_index(name="OrderCount")
          clv_chunk = revenue_chunk.merge(order_count_chunk, on="Customer ID", how="left").fillna(0)
          clv_chunk["AvgPurchaseValue"] = clv_chunk["TotalRevenue"] / clv_chunk["OrderCount"] if clv_chunk['OrderCount'].sum() > 0 else 0
          clv_chunk["CLV"] = clv_chunk["AvgPurchaseValue"] * clv_chunk["OrderCount"] / (3 * 365)
          clv_chunk = clv_chunk[['Customer ID', 'CLV']]
          yield clv_chunk
          del revenue_chunk, order_count_chunk, clv_chunk, chunk
          gc.collect()
        else:
          print("WARNING: Customer ID or Total Revenue column missing in chunk")
          yield pd.DataFrame()
          del chunk
          gc.collect()

def calculate_avg_unit_cost(data_iterator: Iterator[pd.DataFrame]) -> Iterator[pd.DataFrame]:
    """Calculates average unit cost and yields chunks."""

    for i, chunk in enumerate(data_iterator):
        print(f"  - calculate_avg_unit_cost: Processing Chunk {i + 1}")
        if 'Customer ID' in chunk and 'Total Revenue' in chunk and 'Quantity' in chunk:
          avg_unit_cost_chunk = chunk.groupby("Customer ID").apply(
              lambda x: x["Total Revenue"].sum() / x["Quantity"].sum() if x["Quantity"].sum() > 0 else 0
          ).reset_index(name="AvgUnitCost")
          yield avg_unit_cost_chunk
          del avg_unit_cost_chunk, chunk
          gc.collect()
        else:
          print("WARNING: Customer ID or Total Revenue or Quantity column missing in chunk")
          yield pd.DataFrame()
          del chunk
          gc.collect()

def calculate_age(data_iterator: Iterator[pd.DataFrame]) -> Iterator[pd.DataFrame]:
    """Calculates customer age and yields chunks."""

    for i, chunk in enumerate(data_iterator):
        print(f"  - calculate_age: Processing Chunk {i + 1}")
        if 'Customer ID' in chunk and 'Customer_BirthDate' in chunk and 'Order_Date' in chunk:
          age_chunk = chunk[["Customer ID", "Customer_BirthDate", 'Order_Date']].drop_duplicates()
          age_chunk['Latest_Date'] = chunk['Order_Date'].max()
          age_chunk["Customer_Age"] = (
              (age_chunk["Latest_Date"] - age_chunk["Customer_BirthDate"]).dt.days
          ) / 365
          age_chunk = age_chunk[['Customer ID', 'Customer_Age']]
          yield age_chunk
          del age_chunk, chunk
          gc.collect()
        else:
          print("WARNING: Customer ID or Customer_BirthDate or Order_Date column missing in chunk")
          yield pd.DataFrame()
          del chunk
          gc.collect()

def concatenate_features(feature_generators: List[Iterator[pd.DataFrame]]) -> pd.DataFrame:
    """Concatenates features from generators, minimizing memory."""

    final_features = pd.DataFrame()
    first = True
    for generator in feature_generators:
        for chunk in generator:
          if first:
            final_features = chunk
            first = False
          else:
            if 'Customer ID' in final_features.columns and 'Customer ID' in chunk.columns:
              final_features = final_features.merge(chunk, on="Customer ID", how="outer")
            else:
              final_features = pd.concat([final_features, chunk], axis=1)
          del chunk
          gc.collect()
        gc.collect()
    return final_features

def create_features(data_iterator: Iterator[pd.DataFrame]) -> pd.DataFrame:
    """Creates customer segmentation features from chunked data."""

    start_time = datetime.datetime.now()
    print(f"3. create_features started at: {start_time}")

    # Create a pipeline of feature calculation generators
    feature_generators = [
        calculate_frequency(data_iterator),
        calculate_recency(load_and_explore_data(data_url, chunksize=10**4)), # Reset iterator
        calculate_clv(load_and_explore_data(data_url, chunksize=10**4)),
        calculate_avg_unit_cost(load_and_explore_data(data_url, chunksize=10**4)),
        calculate_age(load_and_explore_data(data_url, chunksize=10**4)),
    ]

    # Concatenate features from generators
    feature_df = concatenate_features(feature_generators)

    print("\nFeature DataFrame:\n", feature_df.head())
    print("\nFeature DataFrame Info:\n", feature_df.info(memory_usage="deep"))

    end_time = datetime.datetime.now()
    print(f"3. create_features ended at: {end_time}, duration: {end_time - start_time}")

    gc.collect()
    return feature_df

def explore_engineered_features(feature_df: pd.DataFrame):
    """Explores engineered features."""

    start_time = datetime.datetime.now()
    print(f"4. explore_engineered_features started at: {start_time}")

    print("\n--- 4. EDA of Engineered Features ---")
    if not feature_df.empty:
      print("\nDescriptive Statistics:\n", feature_df.describe())

      # Histograms
      feature_df.hist(figsize=(12, 10))
      plt.suptitle("Distribution of Engineered Features", y=1.02)
      plt.tight_layout()
      plt.show()

      # Boxplots
      plt.figure(figsize=(20, 15))
      features_to_plot = ["Frequency", "Recency", "CLV", "AvgUnitCost", "Customer_Age"]
      for i, col in enumerate(features_to_plot):
          plt.subplot(3, 2, i + 1)
          sns.boxplot(y=feature_df[col])
          plt.title(col)
      plt.tight_layout()
      plt.show()
    else:
      print("WARNING: Feature DataFrame is empty")

    end_time = datetime.datetime.now()
    print(
        f"4. explore_engineered_features ended at: {end_time}, duration: {end_time - start_time}"
    )

def scale_features(feature_df: pd.DataFrame, features_to_scale: List[str]):
    """Scales features using StandardScaler."""

    start_time = datetime.datetime.now()
    print(f"5. scale_features started at: {start_time}")
    if not feature_df.empty:
      scaler = StandardScaler()
      feature_df[features_to_scale] = scaler.fit_transform(
          feature_df[features_to_scale]
      )
    else:
      print("WARNING: Feature DataFrame is empty")

    end_time = datetime.datetime.now()
    print(f"5. scale_features ended at: {end_time}, duration: {end_time - start_time}")
    return feature_df

def find_optimal_k(
    feature_df: pd.DataFrame, features_to_scale: List[str], k_values: List[int]
):
    """Finds optimal number of clusters (k)."""

    start_time = datetime.datetime.now()
    print(f"6. find_optimal_k started at: {start_time}")

    wcss = []
    silhouette_scores = []
    if not feature_df.empty:
      for k in k_values:
          kmeans = KMeans(n_clusters=k, random_state=42, n_init="auto")
          kmeans.fit(feature_df[features_to_scale])
          wcss.append(kmeans.inertia_)
          if k > 2:
              silhouette_scores.append(
                  silhouette_score(feature_df[features_to_scale], kmeans.labels_)
              )

      # Elbow plot
      plt.figure(figsize=(12, 6))
      plt.subplot(1, 2, 1)
      plt.plot(k_values, wcss, marker="o")
      plt.title("Elbow Method for Optimal k")
      plt.xlabel("Number of Clusters (k)")
      plt.ylabel("WCSS")

      # Silhouette plot
      plt.subplot(1, 2, 2)
      plt.plot(k_values[1:], silhouette_scores, marker="o")
      plt.title("Silhouette Score for Optimal k")
      plt.xlabel("Number of Clusters (k)")
      plt.ylabel("Silhouette Score")
      plt.tight_layout()
      plt.show()

      print("\nSilhouette Scores:", silhouette_scores)

      # Dendrogram
      linked = linkage(feature_df[features_to_scale], method="ward")
      plt.figure(figsize=(12, 6))
      dendrogram(
          linked,
          orientation="top",
          labels=feature_df["Customer ID"].tolist() if 'Customer ID' in feature_df else None,
      )
      plt.title("Dendrogram for Hierarchical Clustering")
      plt.xlabel("Customer ID")
      plt.ylabel("Cluster Distance")
      plt.show()
    else:
      print("WARNING: Feature DataFrame is empty")

    end_time = datetime.datetime.now()
    print(f"6. find_optimal_k ended at: {end_time}, duration: {end_time - start_time}")

def perform_kmeans_clustering(
    feature_df: pd.DataFrame, features_to_scale: List[str], optimal_k: int
):
    """Performs K-means clustering."""

    start_time = datetime.datetime.now()
    print(f"7. perform_kmeans_clustering started at: {start_time}")
    if not feature_df.empty:
      kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init="auto")
      feature_df["Cluster"] = kmeans.fit_predict(feature_df[features_to_scale])

      cluster_assignment = feature_df[["Customer ID", "Cluster"]]
      print("\nCluster Assignments:\n", cluster_assignment.head(10))
    else:
      print("WARNING: Feature DataFrame is empty")

    end_time = datetime.datetime.now()
    print(
        f"7. perform_kmeans_clustering ended at: {end_time}, duration: {end_time - start_time}"
    )
    return feature_df

def visualize_cluster_analysis(
    feature_df: pd.DataFrame, features_to_visualize: List[str]
):
    """Visualizes cluster analysis using boxplots."""

    start_time = datetime.datetime.now()
    print(f"8. visualize_cluster_analysis started at: {start_time}")
    if not feature_df.empty:
      fig, axes = plt.subplots(
          1, len(features_to_visualize), figsize=(25, 5), sharey=False
      )
      fig.suptitle("Cluster Analysis: Feature Distributions", fontsize=16)

      for i, feature in enumerate(features_to_visualize):
          sns.boxplot(ax=axes[i], x="Cluster", y=feature, data=feature_df)
          axes[i].set_title(feature, fontsize=12)
          axes[i].set_xlabel("Cluster", fontsize=10)
          axes[i].set_ylabel("Value", fontsize=10)
      plt.tight_layout()
      plt.show()
    else:
      print("WARNING: Feature DataFrame is empty")

    end_time = datetime.datetime.now()
    print(
        f"8. visualize_cluster_analysis ended at: {end_time}, duration: {end_time - start_time}"
    )

def perform_dimensionality_reduction(
    feature_df: pd.DataFrame, features_to_scale: List[str]
):
    """Performs dimensionality reduction using PCA and t-SNE."""

    start_time = datetime.datetime.now()
    print(f"9. perform_dimensionality_reduction started at: {start_time}")
    if not feature_df.empty:
      pca = PCA(n_components=2, random_state=42)
      feature_df["PCA1"], feature_df["PCA2"] = pca.fit_transform(
          feature_df[features_to_scale]
      ).T
      print("Performed PCA.")
      gc.collect()

      tsne = TSNE(
          n_components=2, n_iter=250, n_jobs=-1, init="pca", random_state=42
      )
      feature_df["TSNE1"], feature_df["TSNE2"] = tsne.fit_transform(
          feature_df[features_to_scale]
      ).T
      print("Performed t-SNE.")
      gc.collect()
    else:
      print("WARNING: Feature DataFrame is empty")

    end_time = datetime.datetime.now()
    print(
        f"9. perform_dimensionality_reduction ended at: {end_time}, duration: {end_time - start_time}"
    )
    return feature_df

def visualize_clusters_2d(feature_df: pd.DataFrame):
    """Visualizes clusters in 2D using PCA and t-SNE."""

    start_time = datetime.datetime.now()
    print(f"10. visualize_clusters_2d started at: {start_time}")
    if not feature_df.empty:
      # PCA plot
      plt.figure(figsize=(12, 6))
      plt.subplot(1, 2, 1)
      sns.scatterplot(
          x="PCA1", y="PCA2", hue="Cluster", data=feature_df, palette="viridis"
      )
      plt.title("PCA Visualization of Clusters")
      plt.xlabel("PCA Component 1")
      plt.ylabel("PCA Component 2")

      # t-SNE plot
      plt.subplot(1, 2, 2)
      sns.scatterplot(
          x="TSNE1", y="TSNE2", hue="Cluster", data=feature_df, palette="viridis"
      )
      plt.title("t-SNE Visualization of Clusters")
      plt.xlabel("t-SNE Component 1")
      plt.ylabel("t-SNE Component 2")

      plt.tight_layout()
      plt.show()
    else:
      print("WARNING: Feature DataFrame is empty")

    end_time = datetime.datetime.now()
    print(
        f"10. visualize_clusters_2d ended at: {end_time}, duration: {end_time - start_time}"
    )

def main():
    global data_url
    start_time_main = datetime.datetime.now()
    print(f"Main Execution started at: {start_time_main}")

    data_url = "https://github.com/FourthRevGitHub/CAM_DS_Public-Data-Sets/raw/main/Course%201/Week%206/CUSTOMERS_CLEAN.zip"
    data_iterator = load_and_explore_data(data_url)
    data_iterator = clean_data(data_iterator)
    feature_df = create_features(data_iterator)
    if feature_df.empty:
      print("ERROR: Feature engineering failed. Exiting")
      return
    explore_engineered_features(feature_df)
    features_to_scale = ["Frequency", "Recency", "CLV", "AvgUnitCost", "Customer_Age"]
    feature_df = scale_features(feature_df, features_to_scale)
    k_values = range(2, 11)
    find_optimal_k(feature_df, features_to_scale, k_values)
    optimal_k = 4
    feature_df = perform_kmeans_clustering(
        feature_df, features_to_scale, optimal_k
    )
    features_to_visualize = [
        "Frequency",
        "Recency",
        "CLV",
        "AvgUnitCost",
        "Customer_Age",
    ]
    visualize_cluster_analysis(feature_df, features_to_visualize)
    feature_df = perform_dimensionality_reduction(feature_df, features_to_scale)
    visualize_clusters_2d(feature_df)

    if not feature_df.empty:
      feature_df.to_csv("customer_segmentation_data.csv", index=False)
      print("\nFinal DataFrame saved to 'customer_segmentation_data.csv'")
    else:
      print("WARNING: Feature DataFrame is empty. Not saving")

    end_time_main = datetime.datetime.now()
    print(
        f"Main Execution ended at: {end_time_main}, duration: {end_time_main - start_time_main}"
    )

if __name__ == "__main__":
    main()
```
