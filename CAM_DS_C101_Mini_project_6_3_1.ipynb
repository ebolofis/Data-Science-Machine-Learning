```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import datetime
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from typing import List

def load_and_explore_data(url: str) -> pd.DataFrame:
    start_time = datetime.datetime.now()
    print(f"1. load_and_explore_data started at: {start_time}")
    data = pd.read_csv(url, low_memory=False)
    print("\nFirst 5 rows:\n", data.head())
    print("\nDataFrame Info:\n", data.info())
    print("\nDescriptive Statistics:\n", data.describe())
    end_time = datetime.datetime.now()
    print(f"1. load_and_explore_data ended at: {end_time}, duration: {end_time - start_time}")
    return data

def clean_currency_values(series: pd.Series) -> pd.Series:
    return series.astype(str).str.replace(r"[$,()]", "", regex=True).astype(float)

def cap_outliers_np(series: pd.Series, lower_percentile: float = 0.05, upper_percentile: float = 0.95) -> pd.Series:
    lower_bound = np.percentile(series, lower_percentile * 100)
    upper_bound = np.percentile(series, upper_percentile * 100)
    return np.clip(series, lower_bound, upper_bound)

def clean_data(data: pd.DataFrame) -> pd.DataFrame:
    start_time = datetime.datetime.now()
    print(f"2. clean_data started at: {start_time}")
    currency_cols = ["Total Revenue", "Unit Cost", "Profit"]
    for col in currency_cols:
        data[col] = clean_currency_values(data[col])
    date_cols = ["Order_Date", "Delivery_Date", "Customer_BirthDate"]
    for col in date_cols:
        data[col] = pd.to_datetime(data[col], format="%d%b%Y", errors="coerce")
        data.dropna(subset=[col], inplace=True)
        if data.empty:
            print("ERROR: DataFrame is empty after date parsing.")
            return pd.DataFrame()
    data.dropna(subset=["City", "Postal_Code"], inplace=True)
    data.drop(columns=["State_Province"], inplace=True)
    data.drop_duplicates(inplace=True)
    numerical_cols = data.select_dtypes(include=[np.number]).columns.tolist()
    numerical_cols_to_cap = ["Quantity", "Total Revenue", "Unit Cost", "Profit", "Days to Delivery"]
    for col in numerical_cols_to_cap:
        data[col] = cap_outliers_np(data[col])
    end_time = datetime.datetime.now()
    print(f"2. clean_data ended at: {end_time}, duration: {end_time - start_time}")
    return data

def create_features(data: pd.DataFrame) -> pd.DataFrame:
    start_time = datetime.datetime.now()
    print(f"3. create_features started at: {start_time}")
    agg_data = data.groupby("Customer ID")
    frequency_df = agg_data.size().reset_index(name="Frequency")
    latest_date = data["Order_Date"].max()
    recency_df = agg_data["Order_Date"].max().reset_index()
    recency_df["Recency"] = (latest_date - recency_df["Order_Date"]).dt.days
    recency_df.drop(columns=["Order_Date"], inplace=True)
    revenue_per_customer = agg_data["Total Revenue"].sum().reset_index()
    order_counts_per_customer = agg_data.size().reset_index(name="OrderCount")
    clv_df = revenue_per_customer.merge(order_counts_per_customer, on="Customer ID")
    clv_df["AvgPurchaseValue"] = clv_df["Total Revenue"] / clv_df["OrderCount"]
    clv_df["CLV"] = clv_df["AvgPurchaseValue"] * clv_df["OrderCount"] / (3 * 365)
    clv_df.drop(columns=["AvgPurchaseValue", "OrderCount"], inplace=True)
    avg_unit_cost_df = agg_data.apply(lambda x: x["Total Revenue"].sum() / x["Quantity"].sum()).reset_index(name="AvgUnitCost")
    age_df = data[["Customer ID", "Customer_BirthDate"]].drop_duplicates()
    age_df["Customer_Age"] = ((latest_date - age_df["Customer_BirthDate"]).dt.days / 365)
    age_df.drop(columns=["Customer_BirthDate"], inplace=True)
    feature_df = frequency_df.merge(recency_df, on="Customer ID", how="left")
    feature_df = feature_df.merge(clv_df, on="Customer ID", how="left")
    feature_df = feature_df.merge(avg_unit_cost_df, on="Customer ID", how="left")
    feature_df = feature_df.merge(age_df, on="Customer ID", how="left")
    print("\nFeature DataFrame:\n", feature_df.head())
    print("\nFeature DataFrame Info:\n", feature_df.info())
    end_time = datetime.datetime.now()
    print(f"3. create_features ended at: {end_time}, duration: {end_time - start_time}")
    return feature_df

def explore_engineered_features(feature_df: pd.DataFrame):
    start_time = datetime.datetime.now()
    print(f"4. explore_engineered_features started at: {start_time}")
    print("\nDescriptive Statistics:\n", feature_df.describe())
    end_time = datetime.datetime.now()
    print(f"4. explore_engineered_features ended at: {end_time}, duration: {end_time - start_time}")

def scale_features(feature_df: pd.DataFrame, features_to_scale: List[str]) -> pd.DataFrame:
    start_time = datetime.datetime.now()
    print(f"5. scale_features started at: {start_time}")
    scaler = StandardScaler()
    feature_df[features_to_scale] = scaler.fit_transform(feature_df[features_to_scale])
    end_time = datetime.datetime.now()
    print(f"5. scale_features ended at: {end_time}, duration: {end_time - start_time}")
    return feature_df

def find_optimal_k(feature_df: pd.DataFrame, features_to_scale: List[str], k_values: List[int]):
    start_time = datetime.datetime.now()
    print(f"6. find_optimal_k started at: {start_time}")
    wcss = []
    silhouette_scores = []
    for k in k_values:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init="auto")
        kmeans.fit(feature_df[features_to_scale])
        wcss.append(kmeans.inertia_)
        if k > 2:
            silhouette_scores.append(silhouette_score(feature_df[features_to_scale], kmeans.labels_))
    print("\nSilhouette Scores:", silhouette_scores)
    end_time = datetime.datetime.now()
    print(f"6. find_optimal_k ended at: {end_time}, duration: {end_time - start_time}")

def perform_kmeans_clustering(feature_df: pd.DataFrame, features_to_scale: List[str], optimal_k: int) -> pd.DataFrame:
    start_time = datetime.datetime.now()
    print(f"7. perform_kmeans_clustering started at: {start_time}")
    kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init="auto")
    feature_df["Cluster"] = kmeans.fit_predict(feature_df[features_to_scale])
    end_time = datetime.datetime.now()
    print(f"7. perform_kmeans_clustering ended at: {end_time}, duration: {end_time - start_time}")
    return feature_df

def visualize_cluster_analysis(feature_df: pd.DataFrame, features_to_visualize: List[str]):
    start_time = datetime.datetime.now()
    print(f"8. visualize_cluster_analysis started at: {start_time}")
    end_time = datetime.datetime.now()
    print(f"8. visualize_cluster_analysis ended at: {end_time}, duration: {end_time - start_time}")

def perform_dimensionality_reduction(
    feature_df: pd.DataFrame, features_to_scale: List[str]
) -> pd.DataFrame:
    """Performs dimensionality reduction using PCA and t-SNE."""

    start_time = datetime.datetime.now()
    print(f"9. perform_dimensionality_reduction started at: {start_time}")

    pca = PCA(n_components=2, random_state=42)
    feature_df["PCA1"], feature_df["PCA2"] = pca.fit_transform(
        feature_df[features_to_scale]
    ).T

    tsne = TSNE(
        n_components=2,
        n_iter=250,  # Reduced n_iter for speed
        n_jobs=-1,
        init='pca',
        random_state=42
    )
    feature_df["TSNE1"], feature_df["TSNE2"] = tsne.fit_transform(
        feature_df[features_to_scale]
    ).T

    end_time = datetime.datetime.now()
    print(
        f"9. perform_dimensionality_reduction ended at: {end_time}, duration: {end_time - start_time}"
    )
    return feature_df

def visualize_clusters_2d(feature_df: pd.DataFrame):
    start_time = datetime.datetime.now()
    print(f"10. visualize_clusters_2d started at: {start_time}")
    end_time = datetime.datetime.now()
    print(f"10. visualize_clusters_2d ended at: {end_time}, duration: {end_time - start_time}")

def main():
    start_time_main = datetime.datetime.now()
    print(f"Main Execution started at: {start_time_main}")
    data_url = "https://github.com/FourthRevGitHub/CAM_DS_Public-Data-Sets/raw/main/Course%201/Week%206/CUSTOMERS_CLEAN.zip"
    data = load_and_explore_data(data_url)
    if data.empty:
        print("Error: Data loading failed. Exiting.")
        return
    data = clean_data(data)
    if data.empty:
        print("Error: Data cleaning failed. Exiting.")
        return
    feature_df = create_features(data)
    explore_engineered_features(feature_df)
    features_to_scale = ["Frequency", "Recency", "CLV", "AvgUnitCost", "Customer_Age"]
    feature_df = scale_features(feature_df, features_to_scale)
    k_values = range(2, 11)
    find_optimal_k(feature_df, features_to_scale, k_values)
    optimal_k = 4
    feature_df = perform_kmeans_clustering(feature_df, features_to_scale, optimal_k)
    features_to_visualize = ["Frequency", "Recency", "CLV", "AvgUnitCost", "Customer_Age"]
    visualize_cluster_analysis(feature_df, features_to_visualize)
    feature_df = perform_dimensionality_reduction(feature_df, features_to_scale)
    visualize_clusters_2d(feature_df)
    feature_df.to_csv("customer_segmentation_data.csv", index=False)
    print("\nFinal DataFrame saved to 'customer_segmentation_data.csv'")
    end_time_main = datetime.datetime.now()
    print(f"Main Execution ended at: {end_time_main}, duration: {end_time_main - start_time_main}")

if __name__ == "__main__":
    main()
```
