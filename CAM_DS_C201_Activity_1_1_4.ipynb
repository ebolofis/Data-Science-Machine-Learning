Scenario 1: Customer churn prediction

Classification. This is a classification problem because the output variable (churn) is categorical. Customers will either churn (yes) or not churn (no). The goal is to classify customers into one of these two categories.

Scenario 2: Sales forecasting

Regression. This is a regression problem because the output variable (sales volume) is continuous. The company wants to predict a numerical value for next month's sales.

Scenario 3: Customer lifetime value prediction

Regression. This is a regression problem because the output variable (lifetime value) is continuous. The company wants to predict a numerical value representing the total future revenue expected from a customer.

Scenario 4: Credit score rating

Classification. This is a classification problem because the output variable (creditworthiness) is categorical, even though the categories are represented by numbers (1, 2, 3, 4). The financial institution wants to assign customers to one of these distinct creditworthiness categories.

Scenario 5: Predicting customer satisfaction levels

Regression. This is a regression problem because the output variable (customer satisfaction) is on a continuous scale from 1 to 5. Even though the scale has discrete values, it represents a range, and the prediction aims to estimate a specific point within that range. Treating it as regression allows for predicting values like 3.2 or 4.8, reflecting varying degrees of satisfaction.

Scenario 6: Age prediction

Classification. This is a classification problem because the output variable (age group) is categorical. The company wants to classify users into predefined age brackets (e.g., 18-24, 25-34), which are distinct categories.
