```python
# Necessary Imports (Add these at the top of your script)
import pandas as pd
import numpy as np
import datetime
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, QuantileTransformer, PowerTransformer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from scipy.cluster.hierarchy import dendrogram, linkage
import gc # Garbage collector
import traceback # For detailed error reporting
import sys # For checking python version or interactivity
from collections.abc import Iterator # Correct import for Python 3.9+
from typing import List, Dict, Tuple, Optional # Keep other typing imports if needed

# --- Helper Functions (Example - Implement as needed) ---
def clean_currency_values(series: pd.Series) -> pd.Series:
    """
    Example helper function to clean currency strings.
    Removes symbols like '$', ',', converts to numeric.
    Handles potential errors by returning NaN.
    """
    if pd.api.types.is_numeric_dtype(series):
        return series # Already numeric
    try:
        # Attempt cleaning common currency formats
        cleaned = series.astype(str).str.replace(r'[$,]', '', regex=True)
        return pd.to_numeric(cleaned, errors='coerce')
    except Exception:
        # Fallback if unexpected error occurs
        return pd.to_numeric(series, errors='coerce')

def cap_outliers_np(series: pd.Series, lower_quantile=0.01, upper_quantile=0.99) -> pd.Series:
    """
    Caps outliers in a pandas Series using quantiles (NumPy implementation).
    Handles NaN values gracefully.
    """
    # Work on a copy to avoid modifying the original Series passed in
    series_copy = series.copy()
    # Get non-NaN values for quantile calculation
    valid_data = series_copy.dropna()
    if valid_data.empty:
        return series_copy # Return original if all are NaN

    # Calculate quantiles on valid data
    lower_bound = np.quantile(valid_data, lower_quantile)
    upper_bound = np.quantile(valid_data, upper_quantile)

    # Apply capping only to non-NaN values
    # Use np.where for efficient conditional assignment
    series_copy.loc[valid_data.index] = np.where(
        valid_data < lower_bound,
        lower_bound,
        np.where(
            valid_data > upper_bound,
            upper_bound,
            valid_data # Keep original value if within bounds
        )
    )
    return series_copy

# --- Configuration  ---

data_url = "https://github.com/FourthRevGitHub/CAM_DS_Public-Data-Sets/raw/main/Course%201/Week%206/CUSTOMERS_CLEAN.zip" # Replace with the actual URL or file path
# Example local path: data_url = "data/customer_data.csv"


chunk_size = 100000 # Adjust chunk size based on memory
features_to_scale_and_cluster = ["Frequency", "Recency", "CLV", "AvgUnitCost", "Customer_Age"]
k_range_to_test = range(2, 9) # Example: Test k from 2 to 8
# optimal_k = 4 # Set this after reviewing plots, or use input prompt
output_filename = "customer_segments_final.csv"
optimal_k = None # Initialize, will be set by user input or default

# --------------------------------------------------------------------------
# Function Definitions
# --------------------------------------------------------------------------

# --- 1. Data Loading and Initial Exploration Function (MODIFIED + Comments Restored) ---
def load_and_explore_data(url: str, chunksize=10**4) -> Iterator[pd.DataFrame]:
    """
    Loads data in chunks, optimizes data types,
    and performs initial exploration, including explicit reporting
    on missing values, duplicates, and potential outliers for the first chunk.
    """
    start_time = datetime.datetime.now()
    print(f"--- 1. load_and_explore_data started at: {start_time} ---")

    try:
        # Attempt to read the CSV with chunksize
        data_chunks = pd.read_csv(url, low_memory=False, chunksize=chunksize)
    except FileNotFoundError:
         print(f"!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!")
         print(f"ERROR: File not found at the specified path: {url}")
         print(f"Please ensure the 'data_url' variable in the Configuration section")
         print(f"is set correctly to the path or URL of your CSV file.")
         print(f"!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!")
         yield pd.DataFrame() # Return empty DataFrame on error
         return # Terminate the generator
    except Exception as e:
        print(f"ERROR: Failed to read CSV from {url}. Error: {e}")
        yield pd.DataFrame() # Return empty DataFrame on error
        return # Terminate the generator

    # Possible date formats
    date_formats = ["%Y-%m-%d", "%m/%d/%Y", "%d/%m/%Y", "%Y/%m/%d", "%d%b%Y", "%b %d, %Y", "%B %d, %Y", '%Y-%m-%dT%H:%M:%S.%f']

    total_rows = 0
    initial_exploration_done = False # Flag to ensure initial exploration runs only once

    for i, chunk in enumerate(data_chunks):
        print(f"\n--- Processing Chunk {i + 1} ---")
        chunk_start_time = datetime.datetime.now()
        current_chunk_rows = len(chunk)
        total_rows += current_chunk_rows

        # --- Basic Check and Preparation of Customer ID ---
        if 'Customer ID' not in chunk.columns:
            print(f"  WARNING: 'Customer ID' not found in chunk {i+1}. Skipping chunk.")
            del chunk
            gc.collect()
            continue # Skip chunk if Customer ID is missing

        try:
            # Convert Customer ID to string for consistency, handle NaN
            if pd.api.types.is_numeric_dtype(chunk['Customer ID']):
                # Fill NaN with a placeholder value before converting to int
                chunk['Customer ID'] = chunk['Customer ID'].fillna(-1).astype(int).astype(str)
            else:
                # Directly convert to string if not already numeric
                chunk['Customer ID'] = chunk['Customer ID'].fillna('Unknown').astype(str)
        except Exception as e:
            print(f"  WARNING: Could not reliably convert 'Customer ID' to string in chunk {i+1}. Error: {e}. Skipping chunk.")
            del chunk
            gc.collect()
            continue

        # --- Data Type Optimization & Date Handling (Initial Pass) ---
        print("  - Performing initial type optimization...")
        for col in chunk.columns:
            if col == 'Customer ID': # Already handled
                continue
            try:
                # Optimize numeric types
                if pd.api.types.is_numeric_dtype(chunk[col]):
                    # Downcast for memory saving
                    chunk[col] = pd.to_numeric(chunk[col], downcast="integer")
                    chunk[col] = pd.to_numeric(chunk[col], downcast="float")
                # Try converting object columns that might be numeric
                elif pd.api.types.is_object_dtype(chunk[col]):
                    # Attempt conversion BEFORE checking for dates below
                    numeric_attempt = pd.to_numeric(chunk[col], errors='coerce')
                    if not numeric_attempt.isnull().all(): # Check if conversion worked for some
                        # If it looks numeric, try converting
                        chunk[col] = pd.to_numeric(chunk[col], errors='ignore') # Apply if potentially numeric

                # Date handling after potential numeric conversion
                # Check if the column name suggests it's a date
                if col in ["Order_Date", "Delivery_Date", "Customer_BirthDate"]:
                    # Only parse if not already datetime (avoids warnings/errors)
                    if not pd.api.types.is_datetime64_any_dtype(chunk[col]):
                        # Attempt standard parsing first
                        chunk[col] = pd.to_datetime(chunk[col], errors='coerce')
                        # If standard parsing failed (NaT exists), try specific formats
                        if chunk[col].isnull().any():
                            for fmt in date_formats:
                                # Attempt format, coerce errors to NaT
                                try_format = pd.to_datetime(chunk[col].astype(str), format=fmt, errors='coerce')
                                # Use combine_first: Fill NaNs in original with successfully parsed values from try_format
                                chunk[col] = chunk[col].combine_first(try_format)
            except Exception as e:
                # Catch potential errors during optimization (e.g., mixed types)
                print(f"  - Warning: Error during initial type optimization for column '{col}': {e}")


        # --- Initial Exploration & Reporting (First chunk ONLY) ---
        # This block runs only once on the first chunk to give an initial overview
        if not initial_exploration_done:
            print("\n=== Initial Exploration & Data Quality Report (Chunk 1) ===")
            print(f"Analysing first chunk with {current_chunk_rows} rows.")

            # 1. Report Missing Values for the first chunk
            print("\n1. Missing Value Analysis (Chunk 1):")
            missing_counts = chunk.isnull().sum()
            missing_percent = (missing_counts / current_chunk_rows) * 100
            # Create a DataFrame to display missing counts and percentages nicely
            missing_df = pd.DataFrame({'Missing Count': missing_counts, 'Percentage (%)': missing_percent})
            # Filter to show only columns with missing values and sort
            missing_df = missing_df[missing_df['Missing Count'] > 0].sort_values(by='Percentage (%)', ascending=False)
            if not missing_df.empty:
                print(missing_df)
            else:
                print("  - No missing values found in the first chunk.")

            # 2. Report Duplicate Rows for the first chunk
            print("\n2. Duplicate Row Analysis (Chunk 1):")
            duplicates_count = chunk.duplicated().sum()
            print(f"  - Number of duplicate rows found in the first chunk: {duplicates_count}")

            # 3. Identify Potential Outliers (Visual - First Chunk)
            print("\n3. Outlier Identification (Visual - Chunk 1):")
            # Identify potential numeric columns for boxplots (base numeric + try converting currency/profit)
            potential_numeric_cols = ['Quantity'] # Start with known numeric
            currency_cols_to_check = ['Total Revenue', 'Unit Cost', 'Profit']
            # Use a temporary DataFrame to hold potentially converted numeric data for plotting
            temp_numeric_data = pd.DataFrame(index=chunk.index)

            # Attempt to clean and convert currency/profit columns *just for this visualization*
            for col in currency_cols_to_check:
                if col in chunk.columns:
                    # Apply the cleaning function to a copy of the series
                    cleaned_col = clean_currency_values(chunk[col].copy())
                    # Add to temp df only if conversion resulted in some numbers
                    if not cleaned_col.isnull().all():
                        temp_numeric_data[col] = cleaned_col
                        if col not in potential_numeric_cols: # Add name to list if successful
                            potential_numeric_cols.append(col)

            # Add existing Quantity column if it's numeric
            if 'Quantity' in chunk.columns and pd.api.types.is_numeric_dtype(chunk['Quantity']):
                temp_numeric_data['Quantity'] = chunk['Quantity']
            elif 'Quantity' in potential_numeric_cols: # Remove if it wasn't actually numeric
                potential_numeric_cols.remove('Quantity')

            # Generate boxplots if we found suitable columns
            if not temp_numeric_data.empty and potential_numeric_cols:
                print("  - Generating boxplots for potential numeric columns to identify outliers visually...")
                num_plots = len(potential_numeric_cols) # Use the count of confirmed columns
                num_cols_viz = 3
                num_rows_viz = (num_plots + num_cols_viz - 1) // num_cols_viz
                plt.figure(figsize=(num_cols_viz * 5, num_rows_viz * 4))
                plt.suptitle("Initial Outlier Check (Boxplots - First Chunk)", fontsize=16, y=1.03)
                plot_idx = 0
                for col in potential_numeric_cols: # Iterate through the confirmed list
                    if col in temp_numeric_data.columns: # Double check it's in the temp df
                        plot_idx += 1
                        ax = plt.subplot(num_rows_viz, num_cols_viz, plot_idx)
                        sns.boxplot(y=temp_numeric_data[col].dropna(), ax=ax) # Drop NaNs for plotting
                        ax.set_title(f"{col} (Initial)")
                        ax.set_ylabel("")
                plt.tight_layout(rect=[0, 0.03, 1, 0.97])
                plt.show()
            else:
                print("  - No suitable numeric columns found for initial outlier visualization in the first chunk.")
            # Clean up the temporary DataFrame
            del temp_numeric_data
            gc.collect()

            # 4. Basic Info & Stats (as before)
            print("\n4. Basic Info (Chunk 1):")
            print("First 5 rows:\n", chunk.head())
            print("\nDataFrame Info:\n")
            chunk.info(memory_usage="deep")
            print("\nDescriptive Statistics:\n", chunk.describe(include='all'))
            print("==========================================================")

            initial_exploration_done = True # Ensure this block runs only once

        # --- Yield Chunk for next processing step ---
        print(f"  - Yielding Chunk {i+1} - Shape: {chunk.shape}. Time: {datetime.datetime.now() - chunk_start_time}")
        yield chunk # Return the chunk (potentially with optimized types)
        # Clean up memory for the current chunk
        del chunk
        gc.collect()

    end_time = datetime.datetime.now()
    print(f"\n--- 1. load_and_explore_data finished. Total rows processed: {total_rows}. Duration: {end_time - start_time} ---")


# --- 2. Data Cleaning Function (Comments Restored) ---
def clean_data(data_iterator: Iterator[pd.DataFrame]) -> Iterator[pd.DataFrame]:
    """Cleans data in chunks."""
    start_time = datetime.datetime.now()
    print(f"\n--- 2. clean_data started at: {start_time} ---")

    processed_chunks = 0
    for i, chunk in enumerate(data_iterator):
        print(f"\n--- Cleaning Chunk {i + 1} ---")
        chunk_start_time = datetime.datetime.now()
        initial_rows = len(chunk)
        if chunk.empty:
            print("  - Skipping empty chunk.")
            continue

        # --- Define columns to process ---
        currency_cols = ["Total Revenue", "Unit Cost", "Profit"]
        date_cols = ["Order_Date", "Delivery_Date", "Customer_BirthDate"]
        location_cols_to_drop_na = ["City", "Postal_Code"] # Rows with NaN in these will be dropped
        cols_to_drop = ["State_Province"] # These columns will be removed entirely
        outlier_cols_initial = ["Quantity", "Total Revenue", "Unit Cost", "Profit"] # Base list for outlier capping

        # 2.1. Clean Currency Columns
        print("  - Cleaning currency columns...")
        for col in currency_cols:
            if col in chunk.columns:
                # Apply cleaning function
                chunk[col] = clean_currency_values(chunk[col])
                # Downcast to float32 for memory efficiency after cleaning
                if pd.api.types.is_numeric_dtype(chunk[col]):
                    chunk[col] = pd.to_numeric(chunk[col], downcast="float")

        # 2.2. Ensure Date Columns are Datetime
        # Note: Actual parsing happens in load_and_explore_data. Here we just verify/report.
        print("  - Verifying date columns...")
        for col in date_cols:
            # Check if conversion worked, attempt again if not (shouldn't usually be needed)
            if col in chunk.columns and not pd.api.types.is_datetime64_any_dtype(chunk[col]):
                print(f"  - Warning: Column '{col}' is not datetime in clean_data. Attempting conversion again.")
                chunk[col] = pd.to_datetime(chunk[col], errors='coerce')
            # Report missing counts for info
            if col in chunk.columns:
                print(f"    - Missing values in {col}: {chunk[col].isnull().sum()}")

        # 2.3. Calculate 'Days to Delivery'
        # Calculated here *after* date columns are verified/cleaned
        outlier_cols = outlier_cols_initial.copy() # Start with base list for this chunk
        if 'Order_Date' in chunk.columns and 'Delivery_Date' in chunk.columns:
            # Check if both columns are actually datetime types before calculating
            if pd.api.types.is_datetime64_any_dtype(chunk['Order_Date']) and pd.api.types.is_datetime64_any_dtype(chunk['Delivery_Date']):
                print("  - Calculating 'Days to Delivery'")
                chunk['Days to Delivery'] = (chunk['Delivery_Date'] - chunk['Order_Date']).dt.days
                # Handle cases where delivery might be before order (data error) -> set to 0
                chunk['Days to Delivery'] = chunk['Days to Delivery'].apply(lambda x: x if pd.isna(x) or x >= 0 else 0)
                print(f"    - Missing/Negative values set to NaN/0 in 'Days to Delivery': {chunk['Days to Delivery'].isnull().sum()} NaN, {(chunk['Days to Delivery'] == 0).sum()} zeros")
                # Add this new feature to the list for outlier capping
                if 'Days to Delivery' not in outlier_cols: # Avoid duplicates if run multiple times
                    outlier_cols.append('Days to Delivery')
            else:
                print("  - Warning: Order_Date or Delivery_Date not in datetime format. Cannot calculate 'Days to Delivery'.")
                chunk['Days to Delivery'] = np.nan # Ensure column exists even if calculation fails

        # 2.4. Handle Missing Values & Drop Columns
        # Drop rows with missing values in specified location columns
        if location_cols_to_drop_na:
            print(f"  - Handling missing values in location columns (dropping rows): {location_cols_to_drop_na}")
            cols_exist = [col for col in location_cols_to_drop_na if col in chunk.columns]
            if cols_exist:
                missing_before = chunk[cols_exist].isnull().sum().sum()
                chunk.dropna(subset=cols_exist, inplace=True) # Drop rows where ANY of these are NaN
                missing_removed = missing_before - chunk[cols_exist].isnull().sum().sum()
                print(f"    - Rows removed due to missing values in {cols_exist}: {missing_removed}")

        # Drop specified columns entirely
        if cols_to_drop:
            print(f"  - Dropping specified columns: {cols_to_drop}")
            cols_to_drop_exist = [col for col in cols_to_drop if col in chunk.columns]
            if cols_to_drop_exist:
                chunk.drop(columns=cols_to_drop_exist, inplace=True, errors="ignore")
                print(f"    - Columns dropped: {cols_to_drop_exist}")

        # 2.5. Handle Duplicates
        print("\n  - Handling duplicate rows")
        duplicates_before = chunk.duplicated().sum()
        if duplicates_before > 0:
            chunk.drop_duplicates(inplace=True)
            print(f"    - Removed {duplicates_before} duplicate rows.")
        else:
            print("    - No duplicate rows found.")

        # 2.6. Handle Outliers (Apply Capping)
        # Use the updated outlier_cols list (which includes Days to Delivery if created)
        print(f"\n  - Handling outliers in columns: {outlier_cols}")
        for col in outlier_cols:
            if col in chunk.columns:
                # Only cap if column is numeric and not all missing
                if chunk[col].isnull().all(): continue
                if not pd.api.types.is_numeric_dtype(chunk[col]): continue
                # Apply the capping function
                chunk[col] = cap_outliers_np(chunk[col])

        # --- Finish Chunk Cleaning ---
        final_rows = len(chunk)
        print(f"  - Chunk {i+1} cleaning finished. Rows removed: {initial_rows - final_rows}. Final shape: {chunk.shape}. Time: {datetime.datetime.now() - chunk_start_time}")

        # Yield the cleaned chunk if it's not empty
        if not chunk.empty:
            processed_chunks += 1
            yield chunk
        else:
            print(f"  - Chunk {i+1} is empty after cleaning, skipping.")

        # Memory management
        del chunk
        gc.collect()

    end_time = datetime.datetime.now()
    print(f"\n--- 2. clean_data finished. Processed {processed_chunks} non-empty chunks. Duration: {end_time - start_time} ---")


# --- 3. Feature Engineering Function (Comments Restored) ---
def create_features(data_iterator: Iterator[pd.DataFrame]) -> pd.DataFrame:
    """Creates customer segmentation features by processing cleaned data chunks **ONCE**."""
    start_time = datetime.datetime.now()
    print(f"\n--- 3. create_features started at: {start_time} ---")

    # Intermediate storage dictionaries to aggregate data across chunks
    frequency_data = {} # {cust_id: count of orders}
    recency_data = {} # {cust_id: most recent order_date}
    monetary_data = {} # {cust_id: sum of total_revenue}
    # For Avg Unit Cost calculation:
    unit_cost_agg = {} # {cust_id: {'total_revenue': sum, 'total_quantity': sum}}
    # For Age calculation:
    age_agg = {} # {cust_id: {'birth_date': first_valid_date, 'latest_order_date_cust': latest_date}}

    latest_overall_order_date = pd.NaT # Track the latest date across the entire dataset for Recency/Age
    processed_chunks = 0
    total_rows_processed = 0

    # --- Process Chunks and Aggregate Intermediate Data ---
    for i, chunk in enumerate(data_iterator): # Iterate through CLEANED chunks
        print(f"\n--- Feature Engineering: Processing Chunk {i + 1} ---")
        chunk_start_time = datetime.datetime.now()
        if chunk.empty:
            print("  - Skipping empty chunk.")
            continue
        processed_chunks += 1
        total_rows_processed += len(chunk)

        # --- Prerequisite Checks ---
        if 'Customer ID' not in chunk.columns:
            print("  - WARNING: 'Customer ID' missing in chunk. Skipping feature calculation for this chunk.")
            continue
        # Ensure Customer ID is string type for consistent dictionary keys
        if not pd.api.types.is_string_dtype(chunk['Customer ID']):
            chunk['Customer ID'] = chunk['Customer ID'].astype(str)

        # Ensure necessary columns have correct types (should be done by clean_data, but double-check)
        date_cols_check = ['Order_Date', 'Customer_BirthDate']
        for col in date_cols_check:
            if col in chunk.columns and not pd.api.types.is_datetime64_any_dtype(chunk[col]):
                print(f"  - WARNING: Column '{col}' is not datetime. Converting...")
                chunk[col] = pd.to_datetime(chunk[col], errors='coerce')
        numeric_cols_check = ['Total Revenue', 'Quantity', 'Unit Cost'] # Unit Cost used indirectly via AvgUnitCost
        for col in numeric_cols_check:
            if col in chunk.columns and not pd.api.types.is_numeric_dtype(chunk[col]):
                print(f"  - WARNING: Column '{col}' is not numeric. Converting...")
                chunk[col] = pd.to_numeric(chunk[col], errors='coerce')

        # --- Update overall latest date from this chunk ---
        if 'Order_Date' in chunk.columns and chunk['Order_Date'].notna().any():
            chunk_max_date = chunk['Order_Date'].max()
            # Update if it's the first date found or later than the current latest
            if pd.isna(latest_overall_order_date) or chunk_max_date > latest_overall_order_date:
                latest_overall_order_date = chunk_max_date

        # --- Calculate intermediate values per customer FOR THE CURRENT CHUNK ---
        print("  - Aggregating intermediate features from chunk...")

        # Frequency (Order Count): Increment count for each customer found in chunk
        if 'Customer ID' in chunk.columns:
            # Group by Customer ID and count occurrences (size) in this chunk
            chunk_freq = chunk.groupby('Customer ID').size()
            # Update the global frequency count for each customer
            for cust_id, count in chunk_freq.items():
                frequency_data[cust_id] = frequency_data.get(cust_id, 0) + count
            del chunk_freq # Free memory

        # Recency (Last Order Date): Update max order date for each customer
        if 'Customer ID' in chunk.columns and 'Order_Date' in chunk.columns:
            # Find the latest order date *per customer* in the current chunk (ignoring NaT)
            chunk_recency = chunk.loc[chunk['Order_Date'].notna()].groupby('Customer ID')['Order_Date'].max()
            # Update the global last order date if this chunk's date is later
            for cust_id, last_date in chunk_recency.items():
                current_max_date = recency_data.get(cust_id, pd.NaT)
                if pd.isna(current_max_date) or last_date > current_max_date:
                    recency_data[cust_id] = last_date
            del chunk_recency # Free memory

        # Monetary (Total Revenue): Add chunk's revenue to customer's total
        if 'Customer ID' in chunk.columns and 'Total Revenue' in chunk.columns:
            # Sum revenue per customer in this chunk (ignoring NaN revenue)
            chunk_monetary = chunk.loc[chunk['Total Revenue'].notna()].groupby('Customer ID')['Total Revenue'].sum()
            # Add this chunk's sum to the global sum for each customer
            for cust_id, revenue in chunk_monetary.items():
                monetary_data[cust_id] = monetary_data.get(cust_id, 0) + revenue
            del chunk_monetary # Free memory

        # Average Unit Cost (Aggregate Sums): Add chunk's revenue and quantity sums
        if 'Customer ID' in chunk.columns and 'Total Revenue' in chunk.columns and 'Quantity' in chunk.columns:
            # Filter for valid rows (revenue/quantity not NaN, quantity > 0)
            valid_rows = chunk['Total Revenue'].notna() & chunk['Quantity'].notna() & (chunk['Quantity'] > 0)
            # Group by customer and sum revenue and quantity for valid rows
            chunk_unit_cost_sums = chunk.loc[valid_rows].groupby('Customer ID')[['Total Revenue', 'Quantity']].sum()
            # Update the global aggregation dictionary
            for cust_id, data in chunk_unit_cost_sums.iterrows():
                # Get existing sums or initialize if first time seeing customer
                agg_data = unit_cost_agg.get(cust_id, {'total_revenue': 0, 'total_quantity': 0})
                # Add current chunk's sums
                agg_data['total_revenue'] += data['Total Revenue']
                agg_data['total_quantity'] += data['Quantity']
                # Store updated sums
                unit_cost_agg[cust_id] = agg_data
            del chunk_unit_cost_sums, valid_rows # Free memory

        # Age (Birth Date and Latest Order Date): Store first valid birth date and track latest order date
        if 'Customer ID' in chunk.columns and 'Customer_BirthDate' in chunk.columns and 'Order_Date' in chunk.columns:
            # Get relevant columns, drop rows where birthdate or order date is missing for this step
            # drop_duplicates helps if a customer appears multiple times in the chunk with same birthdate
            chunk_age_info = chunk.loc[chunk['Customer_BirthDate'].notna() & chunk['Order_Date'].notna(),
                                       ['Customer ID', 'Customer_BirthDate', 'Order_Date']].drop_duplicates(subset=['Customer ID'])
            # Iterate through unique customer info in the chunk
            for _, row in chunk_age_info.iterrows():
                cust_id = row['Customer ID']
                birth_date = row['Customer_BirthDate']
                order_date = row['Order_Date'] # This customer's latest order date IN THIS CHUNK

                # If customer not seen before, store birth date and this order date
                if cust_id not in age_agg:
                    age_agg[cust_id] = {'birth_date': birth_date, 'latest_order_date_cust': order_date}
                else:
                    # If customer exists, update their latest order date if this chunk's is later
                    current_latest_cust = age_agg[cust_id].get('latest_order_date_cust', pd.NaT)
                    if pd.isna(current_latest_cust) or order_date > current_latest_cust:
                        age_agg[cust_id]['latest_order_date_cust'] = order_date
                    # Ensure birth date is stored (in case first seen row had NaT order_date but valid birth date)
                    if 'birth_date' not in age_agg[cust_id] or pd.isna(age_agg[cust_id]['birth_date']):
                        age_agg[cust_id]['birth_date'] = birth_date
            del chunk_age_info # Free memory

        # --- End of Chunk Processing ---
        del chunk # Free memory of the processed chunk
        gc.collect()
        print(f"  - Finished aggregating chunk {i+1}. Time: {datetime.datetime.now() - chunk_start_time}")

    # --- Post-Chunk Processing: Final Feature Calculation ---
    # This happens only ONCE after all chunks are processed
    if processed_chunks == 0:
        print("ERROR: No non-empty chunks were processed during feature engineering.")
        return pd.DataFrame() # Return empty DataFrame

    print(f"\n--- Aggregating Final Features from {processed_chunks} Chunks ({total_rows_processed} total rows) ---")
    print(f"  - Overall latest order date found: {latest_overall_order_date}")

    # Handle case where no order dates were found at all
    if pd.isna(latest_overall_order_date):
        print("ERROR: Could not determine the latest overall order date. Cannot calculate Recency or Age accurately.")
        print("WARNING: Using current date as fallback for latest_overall_order_date.")
        latest_overall_order_date = pd.Timestamp(datetime.datetime.now().date())
        # Alternatively: return pd.DataFrame() # Option to exit if latest date is crucial

    # --- Combine aggregated data into a final DataFrame ---
    # Get all unique customer IDs encountered
    all_customer_ids = set(frequency_data.keys()) | set(recency_data.keys()) | set(monetary_data.keys()) | set(unit_cost_agg.keys()) | set(age_agg.keys())
    if not all_customer_ids:
        print("ERROR: No customer data found after processing chunks.")
        return pd.DataFrame()

    print(f"  - Found {len(all_customer_ids)} unique customer IDs across all data.")
    # Create DataFrame with Customer ID as index for easier mapping
    feature_df = pd.DataFrame(index=list(all_customer_ids))
    feature_df.index.name = 'Customer ID'

    # Map Frequency (default to 0 if customer had no orders but maybe age info)
    feature_df['Frequency'] = feature_df.index.map(frequency_data).fillna(0).astype(int)

    # Calculate and Map Recency
    # Recency = (latest date in whole dataset) - (customer's latest order date)
    feature_df['Last_Order_Date'] = feature_df.index.map(recency_data) # Map the latest date found for each customer
    feature_df['Recency'] = (latest_overall_order_date - feature_df['Last_Order_Date']).dt.days
    # Handle customers who might be missing an order date (e.g., data errors, only birthdate known)
    if feature_df['Recency'].isnull().any():
        # Fill missing recency: Use a value indicating inactivity (e.g., max observed recency + 1, or a large number)
        max_recency = feature_df['Recency'].max()
        # Use max+1 if max exists, otherwise a default large number
        fill_recency_value = max_recency + 1 if pd.notna(max_recency) else 9999
        print(f"  - Filling {feature_df['Recency'].isnull().sum()} missing Recency values with: {fill_recency_value}")
        feature_df['Recency'].fillna(fill_recency_value, inplace=True)
    # Ensure Recency is integer
    feature_df['Recency'] = feature_df['Recency'].astype(int)

    # Map Total Revenue (Monetary Value)
    feature_df['TotalRevenue'] = feature_df.index.map(monetary_data).fillna(0)

    # Calculate and Map Average Unit Cost from aggregated sums
    def calculate_avg_cost(cust_id):
        """Safely calculates average cost from aggregated sums."""
        data = unit_cost_agg.get(cust_id)
        # Check if data exists for customer and quantity is non-zero
        if data and data.get('total_quantity', 0) > 0:
            # Avoid division by zero
            return data['total_revenue'] / data['total_quantity']
        return 0 # Return 0 if no data or 0 quantity
    feature_df['AvgUnitCost'] = feature_df.index.map(calculate_avg_cost).fillna(0).astype('float32') # Use float32

    # Calculate and Map Customer Age
    def calculate_age(cust_id):
        """Calculates age based on stored birth date and reference date."""
        data = age_agg.get(cust_id)
        # Use the customer's latest order date as reference if available, otherwise the overall latest date
        reference_date = data.get('latest_order_date_cust', latest_overall_order_date) if data else latest_overall_order_date

        # Check if birth date and reference date are valid
        if data and pd.notna(data.get('birth_date')) and pd.notna(reference_date):
            birth_date = data['birth_date']
            # Sanity checks: birth date cannot be after reference date
            if birth_date > reference_date: return np.nan
            # Calculate age in days and then years
            age_days = (reference_date - birth_date).days
            age_years = age_days / 365.25 # Account for leap years
            # Add sanity check for realistic age range (e.g., 0-120 years)
            if 0 <= age_years <= 120:
                return age_years
            else:
                return np.nan # Unrealistic age
        return np.nan # Return NaN if no valid birth date or reference date found

    feature_df['Customer_Age'] = feature_df.index.map(calculate_age)
    # Handle potential NaNs in age (e.g., fill with median/mean)
    if feature_df['Customer_Age'].isnull().any():
        median_age = feature_df['Customer_Age'].median()
        # Only fill if median is calculable (i.e., not all ages are NaN)
        if pd.notna(median_age):
            print(f"  - Filling {feature_df['Customer_Age'].isnull().sum()} missing Customer_Age values with median age: {median_age:.2f}")
            feature_df['Customer_Age'].fillna(median_age, inplace=True)
        else:
            # Fallback if median calculation fails
            print(f"  - Warning: Could not calculate median age. Filling {feature_df['Customer_Age'].isnull().sum()} missing Customer_Age with 0.")
            feature_df['Customer_Age'].fillna(0, inplace=True) # Or consider dropping these customers

    # Calculate CLV (Example: Simplified CLV based on Total Revenue)
    # You might replace this with a more sophisticated CLV calculation
    feature_df['CLV'] = feature_df['TotalRevenue']

    # --- Final Cleanup and Column Selection ---
    # Define the final columns needed
    final_feature_columns = ['Frequency', 'Recency', 'CLV', 'AvgUnitCost', 'Customer_Age']
    # Ensure all expected final columns exist in the DataFrame
    for col in final_feature_columns:
        if col not in feature_df.columns:
            print(f"  - Warning: Final feature column '{col}' was missing. Creating with default values.")
            # Initialize with appropriate default (0 for counts/money, NaN for age/recency if calculation failed)
            feature_df[col] = 0 if col in ['Frequency', 'CLV', 'AvgUnitCost'] else np.nan

    # Select only the final feature columns in the desired order
    feature_df = feature_df[final_feature_columns]

    # Drop rows where any of the final essential features are still NaN
    initial_count = len(feature_df)
    feature_df.dropna(subset=final_feature_columns, inplace=True)
    if len(feature_df) < initial_count:
        print(f"  - Dropped {initial_count - len(feature_df)} rows with NaN values in final features.")

    # Reset index to make 'Customer ID' a column again
    feature_df.reset_index(inplace=True)

    # --- Display Results ---
    print("\nFinal Feature DataFrame Head:\n", feature_df.head())
    print("\nFinal Feature DataFrame Info:\n")
    feature_df.info(memory_usage="deep")
    print("\nFinal Feature DataFrame Description:\n", feature_df.describe())

    end_time = datetime.datetime.now()
    print(f"\n--- 3. create_features ended at: {end_time}. Duration: {end_time - start_time} ---")

    # Error check for empty final DataFrame
    if feature_df.empty:
        print("ERROR: Resulting feature DataFrame is empty after aggregation and cleaning.")

    gc.collect()
    return feature_df


# --- 4. Exploratory Data Analysis (EDA) on Engineered Features Function (Comments Restored) ---
def explore_engineered_features(feature_df: pd.DataFrame):
    """Explores the final engineered features using statistics and visualizations."""
    if feature_df.empty:
        print("WARNING: feature_df is empty in explore_engineered_features. Skipping EDA.")
        return

    start_time = datetime.datetime.now()
    print(f"\n--- 4. explore_engineered_features started at: {start_time} ---")

    # List of features created that we want to explore
    features_to_explore = ["Frequency", "Recency", "CLV", "AvgUnitCost", "Customer_Age"]
    # Ensure these columns actually exist in the DataFrame
    features_present = [col for col in features_to_explore if col in feature_df.columns]
    if not features_present:
        print("  - No specified features found to explore.")
        return

    # Create a copy for EDA to avoid modifying the original df passed to this function
    eda_df = feature_df[features_present].copy()

    # 4.1. Descriptive Statistics
    print("\n4.1. Descriptive Statistics:")
    print(eda_df.describe())

    # Check for infinite values which can break plots and replace them
    has_inf = np.isinf(eda_df).any().any()
    if has_inf:
        print("\nWARNING: Infinite values detected in features. Replacing with NaN before plotting.")
        eda_df = eda_df.replace([np.inf, -np.inf], np.nan) # Replace inf with NaN
        # Drop rows with NaN (including those created from inf) for EDA visualizations
        initial_rows = len(eda_df)
        eda_df.dropna(inplace=True)
        print(f"  - Dropped {initial_rows - len(eda_df)} rows containing NaN/inf values for EDA.")
        if eda_df.empty:
            print("  - DataFrame became empty after handling infinite values. Stopping EDA.")
            return

    # 4.2. Histograms
    print("\n4.2. Histograms:")
    try:
        # Plot histograms for all present features
        eda_df.hist(figsize=(12, 10), bins=30) # Increased bins for more detail
        plt.suptitle("Distribution of Engineered Features", y=1.02)
        plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout
        plt.show()
    except Exception as e:
        print(f"  - Error plotting histograms: {e}")

    # 4.3. Boxplots (Shows distribution and identifies potential outliers visually)
    print("\n4.3. Boxplots:")
    try:
        plt.figure(figsize=(15, 10))
        num_plots = len(features_present)
        num_cols = 3
        num_rows = (num_plots + num_cols - 1) // num_cols
        for i, col in enumerate(features_present):
            plt.subplot(num_rows, num_cols, i + 1)
            plot_data = eda_df[col].dropna() # Ensure no NaNs for plotting
            if plot_data.empty:
                print(f"  - Skipping boxplot for '{col}' due to no valid data.")
                plt.title(f"{col}\n(No valid data)")
                continue
            sns.boxplot(y=plot_data) # Generate boxplot
            plt.title(col)
            plt.ylabel("") # Remove y-axis label for cleaner look
            plt.xlabel("") # Remove x-axis label
        plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout
        plt.suptitle("Boxplots of Engineered Features", y=1.03, fontsize=16) # Add overall title
        plt.show()
    except Exception as e:
        print(f"  - Error plotting boxplots: {e}")

    # 4.4. Correlation Matrix
    print("\n4.4. Correlation Matrix:")
    try:
        # Calculate correlation only on numeric types within the selected features
        numeric_features_df = eda_df.select_dtypes(include=np.number)
        if not numeric_features_df.empty:
            plt.figure(figsize=(8, 6))
            correlation_matrix = numeric_features_df.corr()
            # Plot heatmap
            sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
            plt.title("Correlation Matrix of Features")
            plt.show()
        else:
            print("  - No numeric features available for correlation matrix.")
    except Exception as e:
        print(f"  - Error plotting correlation matrix: {e}")

    # 4.5. Interpretations Placeholder
    # This section should be filled in manually in the notebook based on the outputs above
    print("\n4.5. Interpretations (Placeholder - User Action Needed):")
    # Provide brief statistical summaries to aid interpretation
    for col in features_present:
        if col in eda_df.columns:
            stats = eda_df[col].describe()
            print(f"\n{col}:")
            print(f"  - Mean={stats.loc['mean']:.2f}, Median={stats.loc['50%']:.2f}, StdDev={stats.loc['std']:.2f}")
            print(f"  - Min={stats.loc['min']:.2f}, Max={stats.loc['max']:.2f}, Skew={eda_df[col].skew():.2f}")
            print("  - Interpretation: [Add your interpretation based on stats and plots]")
        else:
            print(f"\n{col}: Not available for interpretation.")

    end_time = datetime.datetime.now()
    print(f"--- 4. explore_engineered_features ended at: {end_time}. Duration: {end_time - start_time} ---")
    gc.collect() # Clean up memory after EDA plots

# --- 5. Scaling Function (Comments Restored) ---
def scale_features(feature_df: pd.DataFrame, features_to_scale: List[str]):
    """Scales specified features using StandardScaler."""
    if feature_df.empty:
        print("WARNING: feature_df is empty in scale_features. Skipping scaling.")
        return feature_df, None # Return empty df and no scaler

    start_time = datetime.datetime.now()
    print(f"\n--- 5. scale_features started at: {start_time} ---")

    # Ensure columns to scale actually exist in the DataFrame
    features_present = [col for col in features_to_scale if col in feature_df.columns]
    if not features_present:
        print("  - No features found to scale.")
        return feature_df, None

    # Create a copy of the slice to avoid SettingWithCopyWarning when imputing/scaling
    data_to_scale = feature_df[features_present].copy()

    # Check for NaN/inf before scaling and impute if necessary
    if data_to_scale.isnull().any().any() or np.isinf(data_to_scale).any().any():
        print("  - WARNING: NaN or infinite values detected before scaling.")
        data_to_scale = data_to_scale.replace([np.inf, -np.inf], np.nan) # Replace inf with NaN
        # Impute NaNs using the median of each column before scaling
        for col in data_to_scale.columns:
            if data_to_scale[col].isnull().any():
                median_val = data_to_scale[col].median()
                print(f"  - Imputing {data_to_scale[col].isnull().sum()} NaNs in '{col}' with median ({median_val:.2f}) before scaling.")
                data_to_scale[col].fillna(median_val, inplace=True) # Impute inplace on the copy

    # Perform Scaling
    print(f"  - Scaling features: {features_present}")
    scaler = StandardScaler()
    # Fit the scaler and transform the data
    scaled_data = scaler.fit_transform(data_to_scale)

    # Create new column names for the scaled features
    scaled_cols = [f"{col}_scaled" for col in features_present]
    # Add the scaled data back to the original DataFrame using the new column names
    # This uses the index alignment between feature_df and data_to_scale (derived from feature_df)
    feature_df[scaled_cols] = scaled_data

    print("  - Scaled features head:\n", feature_df[scaled_cols].head())

    end_time = datetime.datetime.now()
    print(f"--- 5. scale_features ended at: {end_time}. Duration: {end_time - start_time} ---")
    # Return the DataFrame with scaled columns and the fitted scaler object
    return feature_df, scaler


# --- 6. Find Optimal K Function (Comments Restored) ---
#
def find_optimal_k(feature_df: pd.DataFrame, scaled_features_cols: List[str], k_values: range):
    """Finds optimal k using Elbow and Silhouette methods, and plots Dendrogram."""
    if feature_df.empty or not scaled_features_cols:
        print("WARNING: feature_df empty or no scaled features provided. Skipping find_optimal_k.")
        return

    # Ensure scaled columns actually exist in the DataFrame
    actual_scaled_cols = [col for col in scaled_features_cols if col in feature_df.columns]
    if not actual_scaled_cols:
        print(f"ERROR: None of the expected scaled columns found. Cannot find optimal k.")
        return

    start_time = datetime.datetime.now()
    print(f"\n--- 6. find_optimal_k started at: {start_time} ---")
    print(f"  - Testing k values: {list(k_values)}")
    print(f"  - Using features: {actual_scaled_cols}")

    # Prepare data: Use only rows without NaN in the relevant scaled columns
    data_for_k = feature_df[actual_scaled_cols].dropna()
    n_samples = len(data_for_k)
    print(f"  - Using {n_samples} samples for finding k (after dropping NaNs).")

    # Check if enough samples exist for the minimum k value
    if n_samples < min(k_values):
        print(f"  - ERROR: Not enough data points ({n_samples}) to test min k ({min(k_values)}).")
        return
    # Check for remaining infinite values (should have been handled in scaling, but check again)
    if np.isinf(data_for_k).any().any():
        print("  - ERROR: Infinite values detected in data for clustering.")
        return

    # Store results for plotting
    wcss = [] # Within-cluster sum of squares (for Elbow)
    silhouette_scores = [] # For Silhouette analysis

    # Iterate through the specified range of k values
    for k in k_values:
        # Silhouette score requires at least 2 clusters and less than n_samples clusters
        can_calc_silhouette = (k > 1 and k < n_samples)

        print(f"  - Running KMeans for k={k}...")
        try:
            # Initialize KMeans - use 'lloyd' algorithm, set n_init to suppress warnings
            kmeans = KMeans(n_clusters=k, algorithm='lloyd', random_state=42, n_init=10)
            # Fit model
            kmeans.fit(data_for_k)
            # Append inertia (WCSS) for Elbow plot
            wcss.append(kmeans.inertia_)

            # Calculate silhouette score if possible
            if can_calc_silhouette:
                score = silhouette_score(data_for_k, kmeans.labels_)
                silhouette_scores.append(score)
                print(f"    - k={k}, WCSS={kmeans.inertia_:.2f}, Silhouette Score={score:.4f}")
            else:
                # Append NaN placeholder if silhouette cannot be calculated for this k
                if k > 1: silhouette_scores.append(np.nan)
                print(f"    - k={k}, WCSS={kmeans.inertia_:.2f}, Silhouette Score: N/A (k={k}, n_samples={n_samples})")

        except Exception as e:
            # Handle potential errors during KMeans execution
            print(f"  - ERROR running KMeans for k={k}: {e}")
            wcss.append(np.nan) # Append NaN if error occurred
            if k > 1: silhouette_scores.append(np.nan)

    # --- Plotting Results ---
    print("\n  - Plotting Elbow and Silhouette...")
    plt.figure(figsize=(14, 6))

    # Elbow Plot
    ax1 = plt.subplot(1, 2, 1)
    # Get indices where WCSS calculation was successful (not NaN)
    valid_wcss_indices = [i for i, w in enumerate(wcss) if pd.notna(w)]
    if valid_wcss_indices:
        # Get corresponding k values and WCSS scores
        plot_k_wcss = [list(k_values)[i] for i in valid_wcss_indices]
        plot_wcss = [wcss[i] for i in valid_wcss_indices]
        # Plot k vs WCSS
        ax1.plot(plot_k_wcss, plot_wcss, marker="o", linestyle='-')
        ax1.set_xticks(plot_k_wcss) # Ensure ticks match tested k values
    ax1.set_title("Elbow Method for Optimal k")
    ax1.set_xlabel("Number of Clusters (k)")
    ax1.set_ylabel("WCSS (Inertia)")
    ax1.grid(True)

    # Silhouette Plot
    ax2 = plt.subplot(1, 2, 2)
    # k values for silhouette start from k=2
    k_values_for_silhouette = [k for k in k_values if k > 1]
    # Get indices where silhouette calculation was successful (not NaN)
    valid_silhouette_indices = [i for i, s in enumerate(silhouette_scores) if pd.notna(s)]
    if valid_silhouette_indices:
        # Get corresponding k values (k>1) and scores
        plot_k_silhouette = [k_values_for_silhouette[i] for i in valid_silhouette_indices]
        plot_silhouette = [silhouette_scores[i] for i in valid_silhouette_indices]
        # Plot k vs Silhouette Score
        ax2.plot(plot_k_silhouette, plot_silhouette, marker="o", linestyle='-')
        ax2.set_xticks(plot_k_silhouette) # Ensure ticks match tested k values

        # Highlight the maximum score if scores were calculated
        if plot_silhouette:
            max_score = max(plot_silhouette)
            # Find the index of the max score to get the corresponding k
            max_k_index = plot_silhouette.index(max_score)
            max_k = plot_k_silhouette[max_k_index]
            # Add a marker for the max score
            ax2.scatter(max_k, max_score, s=150, c='red', label=f'Max Score: {max_score:.4f} at k={max_k}', zorder=5)
            ax2.legend()

    ax2.set_title("Silhouette Score for Optimal k")
    ax2.set_xlabel("Number of Clusters (k)")
    ax2.set_ylabel("Silhouette Score")
    ax2.grid(True)

    plt.tight_layout()
    plt.show()

    # Print scores clearly for review
    print("\n  - Calculated WCSS:", dict(zip(k_values, [f'{w:.2f}' if pd.notna(w) else 'Error' for w in wcss])))
    print("  - Calculated Silhouette Scores:", dict(zip(k_values_for_silhouette, [f'{s:.4f}' if pd.notna(s) else 'N/A or Error' for s in silhouette_scores])))

    # --- Hierarchical Clustering and Dendrogram (Optional, visual aid) ---
    # This helps visualize potential cluster structures, can be slow/memory intensive
    print("\n  - Performing Hierarchical Clustering for Dendrogram (on a sample if large)...")
    sample_size = min(5000, n_samples) # Limit sample size
    if n_samples == 0:
        print("    - Skipping dendrogram: No data points available.")
    # Take a sample if the dataset is larger than the defined sample size
    elif n_samples > sample_size:
        print(f"    - Using a sample of {sample_size} (out of {n_samples}) for dendrogram.")
        data_sample = data_for_k.sample(sample_size, random_state=42)
    else:
        print(f"    - Using all {n_samples} data points for dendrogram.")
        data_sample = data_for_k

    # Proceed only if data_sample is not empty
    if not data_sample.empty:
        try:
            # Perform hierarchical clustering using Ward linkage
            linked = linkage(data_sample, method='ward')
            # Plot the dendrogram
            plt.figure(figsize=(15, 7))
            dendrogram(linked,
                       orientation='top',
                       no_labels=True, # Don't show labels for individual points (too crowded)
                       distance_sort='descending', # Sort branches by distance
                       show_leaf_counts=True) # Show count of points in collapsed nodes
            plt.title(f'Dendrogram for Hierarchical Clustering (Sample Size: {len(data_sample)})')
            plt.xlabel('Data Points (Sampled)')
            plt.ylabel('Cluster Distance (Ward)')
            plt.grid(axis='y', linestyle='--', alpha=0.6)
            plt.show()
        except Exception as e:
            print(f"  - Error generating dendrogram: {e}")

    end_time = datetime.datetime.now()
    print(f"--- 6. find_optimal_k ended at: {end_time}. Duration: {end_time - start_time} ---")
    gc.collect() # Clean up memory


# --- 7. K-Means Clustering Function (Comments Restored) ---
def perform_kmeans_clustering(feature_df: pd.DataFrame, scaled_features_cols: List[str], optimal_k: int) -> pd.DataFrame:
    """Performs K-means clustering with the chosen k and adds cluster labels."""
    # Basic checks for valid input
    if feature_df.empty or not scaled_features_cols or optimal_k is None:
        print("WARNING: feature_df empty, no scaled features, or optimal_k not set. Skipping clustering.")
        feature_df['Cluster'] = np.nan # Ensure column exists but is NaN
        return feature_df

    # Verify that the specified scaled columns actually exist
    actual_scaled_cols = [col for col in scaled_features_cols if col in feature_df.columns]
    if not actual_scaled_cols:
        print(f"ERROR: None of the expected scaled columns found. Cannot perform clustering.")
        feature_df['Cluster'] = np.nan
        return feature_df

    start_time = datetime.datetime.now()
    print(f"\n--- 7. perform_kmeans_clustering started at: {start_time} (optimal_k={optimal_k}) ---")
    print(f"  - Using features: {actual_scaled_cols}")

    # Prepare data: Use only rows without NaN in the relevant scaled columns for clustering
    # Important: Keep track of the original index to map clusters back correctly!
    data_for_clustering = feature_df[actual_scaled_cols].dropna()
    n_samples_cluster = len(data_for_clustering)

    # Check if enough data points exist for the chosen k
    if n_samples_cluster < optimal_k:
        print(f"  - ERROR: Not enough data points ({n_samples_cluster}) for k={optimal_k} after dropping NaNs.")
        feature_df['Cluster'] = np.nan
        return feature_df
    # Check for infinite values
    if np.isinf(data_for_clustering).any().any():
        print("  - ERROR: Infinite values detected in data for clustering.")
        feature_df['Cluster'] = np.nan
        return feature_df

    print(f"  - Performing KMeans with k={optimal_k} on {n_samples_cluster} data points...")
    try:
        # Initialize and fit KMeans model
        kmeans = KMeans(n_clusters=optimal_k, algorithm='lloyd', random_state=42, n_init=10)
        clusters = kmeans.fit_predict(data_for_clustering) # Predict clusters for the valid data

        # Add cluster labels back to the original DataFrame
        # Initialize 'Cluster' column with NaN first
        feature_df['Cluster'] = np.nan
        # Use the index of `data_for_clustering` to assign labels to correct rows in `feature_df`
        feature_df.loc[data_for_clustering.index, 'Cluster'] = clusters
        # Convert cluster labels to integer type (nullable Int64 allows for NaNs)
        feature_df['Cluster'] = feature_df['Cluster'].astype('Int64')

        # --- Analysis of Clustering Results ---
        print("\n  - Cluster Distribution:")
        # Show counts for each cluster
        print(feature_df['Cluster'].value_counts().sort_index().to_frame(name='Count'))

        # Show sample cluster assignments
        if 'Customer ID' in feature_df.columns:
            # Assuming Customer ID was reset_index earlier
            cluster_assignment = feature_df.loc[data_for_clustering.index, ['Customer ID', 'Cluster']].head(10)
            print("\n  - Cluster Assignments (Sample):\n", cluster_assignment)
        else: # If Customer ID was kept as index
            cluster_assignment = feature_df[['Cluster']].head(10)
            print("\n  - Cluster Assignments (Index):\n", cluster_assignment)

        # Analyze Cluster Centers (these are in the SCALED feature space)
        # Helps understand the defining characteristics of each cluster in relative terms
        print("\n  - Cluster Centers (in Scaled Feature Space):")
        centers_df = pd.DataFrame(kmeans.cluster_centers_, columns=actual_scaled_cols)
        print(centers_df.round(3)) # Round for better readability

    except Exception as e:
        print(f"  - ERROR during KMeans fitting or assignment: {e}")
        feature_df['Cluster'] = np.nan # Assign NaN if clustering failed

    end_time = datetime.datetime.now()
    print(f"--- 7. perform_kmeans_clustering ended at: {end_time}. Duration: {end_time - start_time} ---")
    return feature_df


# --- 8. Cluster Visualization and Analysis Functions (Comments Restored) ---
def visualize_cluster_analysis(feature_df: pd.DataFrame, features_to_visualize: List[str]):
    """Visualizes cluster characteristics using original (unscaled) features."""
    # Check if DataFrame and Cluster column are valid
    if feature_df.empty or 'Cluster' not in feature_df.columns or feature_df['Cluster'].isnull().all():
        print("WARNING: Skipping cluster visualization due to missing data or clusters.")
        return

    start_time = datetime.datetime.now()
    print(f"\n--- 8. visualize_cluster_analysis started at: {start_time} ---")

    # Use original (unscaled) features for easier interpretation
    features_present = [f for f in features_to_visualize if f in feature_df.columns]
    if not features_present:
        print("  - No specified features found in DataFrame for visualization.")
        return

    print(f"  - Visualizing features by cluster: {features_present}")
    # Create a DataFrame containing only rows with valid cluster assignments for plotting
    plot_df = feature_df[feature_df['Cluster'].notna()].copy()
    # Convert Cluster column to categorical type for better plotting behavior (especially legends)
    plot_df['Cluster'] = plot_df['Cluster'].astype('category')

    # --- Boxplots for comparing distributions across clusters ---
    num_features = len(features_present)
    num_cols = 3 # Adjust layout columns as needed
    num_rows = (num_features + num_cols - 1) // num_cols
    plt.figure(figsize=(num_cols * 5, num_rows * 4.5)) # Adjust figure size based on layout
    plt.suptitle("Cluster Analysis: Feature Distributions (Original Scale)", fontsize=16, y=1.03)

    for i, feature in enumerate(features_present):
        ax = plt.subplot(num_rows, num_cols, i + 1)
        # Prepare data for this specific plot: handle potential inf/-inf and drop NaNs
        feature_plot_data = plot_df[[feature, 'Cluster']].replace([np.inf, -np.inf], np.nan).dropna()
        # Skip plotting if no valid data remains for this feature
        if feature_plot_data.empty:
            print(f"  - Skipping boxplot for '{feature}' due to no valid data.")
            ax.set_title(f"{feature}\n(No valid data)")
            continue
        # Create boxplot
        sns.boxplot(ax=ax, x="Cluster", y=feature, data=feature_plot_data, palette="viridis", showfliers=True) # showfliers=True to see outliers
        # Optional: Overlay mean points for comparison
        # mean_points = feature_plot_data.groupby('Cluster')[feature].mean().reset_index()
        # sns.stripplot(ax=ax, x='Cluster', y=feature, data=mean_points,
        #               size=8, color='red', marker='D', label='Mean', jitter=False)

        ax.set_title(f"{feature} by Cluster")
        ax.set_xlabel("Cluster")
        ax.set_ylabel(feature) # Keep feature name as y-label
        ax.grid(axis='y', linestyle='--', alpha=0.7)
        # ax.legend() # Add legend if mean points are plotted

    plt.tight_layout(rect=[0, 0, 1, 0.97]) # Adjust layout
    plt.show()

    # --- Calculate and Print Cluster Summaries (using original scales) ---
    print("\n  - Cluster Summaries (Mean Values - Original Scale):")
    # Group by cluster and calculate mean for the features
    try:
        # Calculate mean for all specified features if possible
        cluster_summary = plot_df.groupby('Cluster')[features_present].mean()
        print(cluster_summary.round(2)) # Round for display
    except TypeError as e:
        # If non-numeric columns were included, calculate only for numeric ones
        print(f"  - Warning: Could not calculate mean summary for all features: {e}")
        numeric_features = plot_df[features_present].select_dtypes(include=np.number).columns.tolist()
        if numeric_features:
            cluster_summary = plot_df.groupby('Cluster')[numeric_features].mean()
            print(cluster_summary.round(2))

    # --- Pairplot (Optional) ---
    # Visualizes relationships between pairs of features, colored by cluster
    # Can be slow; uses sampling for performance.
    print("\n  - Generating Pair Plot for Clusters (Sample)...")
    try:
        # Prepare data, ensuring Cluster column is present and no NaNs in features
        pairplot_data = plot_df[features_present + ['Cluster']].dropna()
        n_pairplot = len(pairplot_data)
        sample_size = min(1000, n_pairplot) # Limit sample size

        if n_pairplot == 0:
            print("    - Skipping pairplot: No valid data points.")
        else:
            # Take sample if dataset is large
            pairplot_sample = pairplot_data.sample(sample_size, random_state=42) if n_pairplot > sample_size else pairplot_data

            # Use only numeric features for the pairplot itself
            numeric_features_pairplot = pairplot_sample.select_dtypes(include=np.number).columns.tolist()
            # Get list of features to plot (exclude the Cluster column itself from vars)
            cols_to_plot = [f for f in numeric_features_pairplot if f != 'Cluster']

            if len(cols_to_plot) > 0: # Check if there are numeric features to plot
                sns.pairplot(pairplot_sample, hue="Cluster", palette="viridis",
                             vars=cols_to_plot, # Specify features to plot
                             diag_kind='kde', # Show density plots on diagonal
                             plot_kws={'alpha': 0.6, 's': 40, 'edgecolor': 'k'}, # Customize scatter
                             corner=True) # Plot only lower triangle for efficiency
                plt.suptitle("Pair Plot of Features by Cluster (Sample)", y=1.02)
                plt.show()
            else:
                print("    - Skipping pairplot: Not enough numeric features to plot.")
    except Exception as e:
        print(f"  - Error generating pairplot: {e}")

    end_time = datetime.datetime.now()
    print(f"--- 8. visualize_cluster_analysis ended at: {end_time}. Duration: {end_time - start_time} ---")
    gc.collect() # Clean up memory

# --- 9. Dimensionality Reduction Function (Comments Restored) ---
def perform_dimensionality_reduction(feature_df: pd.DataFrame, scaled_features_cols: List[str]) -> pd.DataFrame:
    """Performs PCA and t-SNE on scaled features for 2D visualization."""
    # Input validation
    if feature_df.empty or not scaled_features_cols:
        print("WARNING: Skipping dimensionality reduction due to empty data or no features.")
        return feature_df

    # Ensure scaled columns exist
    actual_scaled_cols = [col for col in scaled_features_cols if col in feature_df.columns]
    if not actual_scaled_cols:
        print(f"ERROR: Scaled columns not found. Cannot perform reduction.")
        # Initialize columns anyway to prevent errors later if they are expected
        feature_df["PCA1"] = np.nan; feature_df["PCA2"] = np.nan
        feature_df["TSNE1"] = np.nan; feature_df["TSNE2"] = np.nan
        return feature_df

    start_time = datetime.datetime.now()
    print(f"\n--- 9. perform_dimensionality_reduction started at: {start_time} ---")
    print(f"  - Using features: {actual_scaled_cols}")

    # Prepare data: Drop rows with NaNs in the scaled columns to be used
    data_for_reduction = feature_df[actual_scaled_cols].dropna()
    n_samples_redux = len(data_for_reduction)
    print(f"  - Using {n_samples_redux} samples for reduction (after dropping NaNs).")

    # Check if any data remains
    if n_samples_redux == 0:
        print("  - ERROR: No valid data available for reduction after dropping NaNs.")
        feature_df["PCA1"] = np.nan; feature_df["PCA2"] = np.nan
        feature_df["TSNE1"] = np.nan; feature_df["TSNE2"] = np.nan
        return feature_df
    # Check for infinite values
    if np.isinf(data_for_reduction).any().any():
        print("  - ERROR: Infinite values detected in data for reduction.")
        feature_df["PCA1"] = np.nan; feature_df["PCA2"] = np.nan
        feature_df["TSNE1"] = np.nan; feature_df["TSNE2"] = np.nan
        return feature_df

    # --- PCA ---
    print("  - Performing PCA...")
    try:
        # Initialize PCA to reduce to 2 components
        pca = PCA(n_components=2, random_state=42)
        # Fit PCA on the prepared data and transform it
        pca_result = pca.fit_transform(data_for_reduction)
        # Add PCA results back to the original DataFrame using the index for alignment
        feature_df.loc[data_for_reduction.index, "PCA1"] = pca_result[:, 0]
        feature_df.loc[data_for_reduction.index, "PCA2"] = pca_result[:, 1]
        # Report explained variance
        print(f"    - Explained Variance Ratio (PCA): {pca.explained_variance_ratio_}")
        print(f"    - Total Explained Variance (PCA): {pca.explained_variance_ratio_.sum():.4f}")
    except Exception as e:
        print(f"  - ERROR during PCA: {e}")
        # Ensure columns exist even if PCA fails
        feature_df["PCA1"] = np.nan; feature_df["PCA2"] = np.nan

    # --- t-SNE ---
    print("\n  - Performing t-SNE...")
    # t-SNE parameters often need tuning based on dataset size
    n_tsne_points = n_samples_redux
    # Dynamic perplexity (common heuristic)
    perplexity_value = min(max(5, n_tsne_points // 100), 50) if n_tsne_points > 50 else 5
    # Dynamic iterations (common heuristic, capped)
    n_iter_value = min(max(250, n_tsne_points // 10), 1000)

    # Data for t-SNE is the same as for PCA here
    data_for_tsne = data_for_reduction
    if data_for_tsne.empty:
        print("  - Skipping t-SNE as input data is empty.")
        feature_df["TSNE1"] = np.nan; feature_df["TSNE2"] = np.nan
    else:
        print(f"    - Running t-SNE on {len(data_for_tsne)} points (perplexity={perplexity_value}, n_iter={n_iter_value})...")
        start_tsne_time = datetime.datetime.now()
        try:
            # Choose initialization method ('pca' is often faster/better if applicable)
            init_method = 'pca' if data_for_tsne.shape[0] > data_for_tsne.shape[1] else 'random'
            # Initialize t-SNE
            tsne = TSNE(n_components=2, perplexity=perplexity_value, n_iter=n_iter_value,
                        init=init_method, learning_rate='auto', # auto learning rate is usually good
                        random_state=42, n_jobs=-1) # Use all CPU cores
            # Fit t-SNE and transform the data
            tsne_result = tsne.fit_transform(data_for_tsne)
            # Add t-SNE results back to the original DataFrame using the index
            feature_df.loc[data_for_tsne.index, "TSNE1"] = tsne_result[:, 0]
            feature_df.loc[data_for_tsne.index, "TSNE2"] = tsne_result[:, 1]
            print(f"    - t-SNE finished. Time: {datetime.datetime.now() - start_tsne_time}")
        except Exception as e:
            print(f"  - ERROR during t-SNE: {e}")
            # Ensure columns exist even if t-SNE fails
            feature_df["TSNE1"] = np.nan; feature_df["TSNE2"] = np.nan

    end_time = datetime.datetime.now()
    print(f"--- 9. perform_dimensionality_reduction ended at: {end_time}. Duration: {end_time - start_time} ---")
    return feature_df

# --- 10. Visualize Clusters in 2D Function (Comments Restored) ---
def visualize_clusters_2d(feature_df: pd.DataFrame):
    """Visualizes clusters using PCA and t-SNE results."""
    # Input validation
    if feature_df.empty: print("WARNING: feature_df empty. Skipping 2D visualization."); return
    required_cols = ['Cluster', 'PCA1', 'PCA2', 'TSNE1', 'TSNE2']
    # Check if necessary columns exist and have at least some non-NaN values
    cols_present = {col: col in feature_df.columns and feature_df[col].notna().any() for col in required_cols}
    if not cols_present['Cluster']: print("WARNING: 'Cluster' column missing/empty. Skipping."); return
    has_pca = cols_present['PCA1'] and cols_present['PCA2']
    has_tsne = cols_present['TSNE1'] and cols_present['TSNE2']
    if not has_pca and not has_tsne: print("WARNING: No valid PCA or t-SNE results. Skipping."); return

    start_time = datetime.datetime.now()
    print(f"\n--- 10. visualize_clusters_2d started at: {start_time} ---")

    # Prepare data: use only rows with valid cluster assignments
    plot_data_2d = feature_df[feature_df['Cluster'].notna()].copy()
    # Convert Cluster to categorical for discrete coloring and legend
    plot_data_2d['Cluster'] = plot_data_2d['Cluster'].astype('category')

    # Create figure with two subplots
    plt.figure(figsize=(14, 6.5)) # Adjusted size for better legend placement

    # --- PCA Plot ---
    ax1 = plt.subplot(1, 2, 1)
    if has_pca:
        print("  - Plotting PCA...")
        # Use only data points where PCA results are not NaN
        pca_plot_data = plot_data_2d[['PCA1', 'PCA2', 'Cluster']].dropna()
        if not pca_plot_data.empty:
            sns.scatterplot(ax=ax1, x="PCA1", y="PCA2", hue="Cluster", data=pca_plot_data,
                            palette="viridis", s=50, alpha=0.7, edgecolor='k', linewidth=0.5)
            ax1.set_title("PCA Visualization of Clusters")
            ax1.set_xlabel("PCA Component 1"); ax1.set_ylabel("PCA Component 2")
            ax1.grid(True, linestyle='--', alpha=0.5)
            # Move legend outside the plot area
            ax1.legend(title='Cluster', bbox_to_anchor=(1.05, 1), loc='upper left')
        else:
            ax1.set_title("PCA Visualization (No Valid Data)")
            print("  - No valid PCA data to plot.")
    else:
        ax1.set_title("PCA Visualization (Not Available)")
        print("  - PCA columns not found or contain only NaNs.")

    # --- t-SNE Plot ---
    ax2 = plt.subplot(1, 2, 2)
    if has_tsne:
        print("  - Plotting t-SNE...")
        # Use only data points where t-SNE results are not NaN
        tsne_plot_data = plot_data_2d[['TSNE1', 'TSNE2', 'Cluster']].dropna()
        if not tsne_plot_data.empty:
            sns.scatterplot(ax=ax2, x="TSNE1", y="TSNE2", hue="Cluster", data=tsne_plot_data,
                            palette="viridis", s=50, alpha=0.7, edgecolor='k', linewidth=0.5)
            ax2.set_title("t-SNE Visualization of Clusters")
            ax2.set_xlabel("t-SNE Component 1"); ax2.set_ylabel("t-SNE Component 2")
            ax2.grid(True, linestyle='--', alpha=0.5)
            # Move legend outside the plot area
            ax2.legend(title='Cluster', bbox_to_anchor=(1.05, 1), loc='upper left')
        else:
            ax2.set_title("t-SNE Visualization (No Valid Data)")
            print("  - No valid t-SNE data to plot.")
    else:
        ax2.set_title("t-SNE Visualization (Not Available)")
        print("  - t-SNE columns not found or contain only NaNs.")

    # Adjust layout to prevent overlap and make space for legends
    plt.tight_layout(rect=[0, 0, 0.9, 1]) # Leave space on the right for legends
    plt.show()

    end_time = datetime.datetime.now()
    print(f"--- 10. visualize_clusters_2d ended at: {end_time}. Duration: {end_time - start_time} ---")
    gc.collect() # Clean up memory


# --- Main Execution Pipeline ---
def main():
    """Main execution function to run the customer segmentation pipeline."""
    global optimal_k # Make optimal_k global or manage scope differently if preferred
    start_time_main = datetime.datetime.now()
    print(f"===== Main Execution started at: {start_time_main} =====")

    # --- Pipeline Steps ---
    try:
        # Step 1 & 2: Load and Clean Data (Generators)
        # Cleaning happens chunk-by-chunk as data is iterated by create_features
        print("\n=== Step 1 & 2: Loading and Cleaning Data ===")
        data_iterator = load_and_explore_data(data_url, chunksize=chunk_size)
        cleaned_data_iterator = clean_data(data_iterator)

        # Step 3: Create Features
        print("\n=== Step 3: Creating Features ===")
        # This consumes the cleaned_data_iterator
        feature_df = create_features(cleaned_data_iterator)
        if feature_df.empty:
            print("\nERROR: Feature creation resulted in an empty DataFrame. Exiting.")
            return # Exit if no features created

        # Step 4: Explore Engineered Features
        print("\n=== Step 4: Exploratory Data Analysis ===")
        # Plots will be shown if run in an environment that supports matplotlib display
        explore_engineered_features(feature_df.copy()) # Pass copy to avoid modification

        # Step 5: Scale Features
        print("\n=== Step 5: Scaling Features ===")
        scaled_col_names = [f"{col}_scaled" for col in features_to_scale_and_cluster]
        feature_df, scaler = scale_features(feature_df, features_to_scale_and_cluster)
        if scaler is None:
            print("\nERROR: Feature scaling failed. Exiting.")
            return
        # Get list of scaled columns actually created
        actual_scaled_cols_created = [col for col in scaled_col_names if col in feature_df.columns]
        if not actual_scaled_cols_created:
            print("\nERROR: No scaled feature columns found after scaling. Exiting.")
            return
        print(f"\nUsing scaled columns: {actual_scaled_cols_created}")

        # Step 6: Find Optimal K
        print("\n=== Step 6: Finding Optimal K ===")
        # Plots (Elbow, Silhouette, Dendrogram) will be shown here
        find_optimal_k(feature_df, actual_scaled_cols_created, k_range_to_test)

        # --- USER ACTION REQUIRED ---
        # Prompt user to input the optimal k based on the plots
        print("\n--- User Input Required ---")
        try:
            # Check if running interactively (e.g., terminal)
            import sys
            if sys.stdin.isatty():
                optimal_k_input = input(f">>> Review plots. Enter optimal k from {list(k_range_to_test)}: ")
                optimal_k = int(optimal_k_input) # Assign to global/scoped variable
                if optimal_k not in k_range_to_test:
                    raise ValueError(f"Input k={optimal_k} outside tested range {list(k_range_to_test)}.")
            else:
                # Fallback for non-interactive execution
                print("Non-interactive mode detected. Using default k=4. Modify script to change.")
                optimal_k = 4 # Default value
                if optimal_k not in k_range_to_test:
                    raise ValueError(f"Default k={optimal_k} outside tested range {list(k_range_to_test)}.")

            print(f"\n===> Proceeding with k = {optimal_k} <===")
        except ValueError as e:
            print(f"\nERROR: Invalid input or default value for optimal k. {e}. Exiting.")
            return
        except EOFError:
            # Handle script being run where input() fails
            print("\nERROR: Could not read input (EOFError). Running non-interactively? Set default k or run interactively. Exiting.")
            return

        # Step 7: Perform K-Means Clustering
        print("\n=== Step 7: Performing K-Means Clustering ===")
        feature_df = perform_kmeans_clustering(feature_df, actual_scaled_cols_created, optimal_k)
        # Check if clustering was successful
        if 'Cluster' not in feature_df.columns or feature_df['Cluster'].isnull().all():
            print("\nERROR: K-Means clustering failed or produced no valid clusters. Exiting.")
            return

        # Step 8: Visualize Cluster Analysis
        print("\n=== Step 8: Visualizing Cluster Analysis ===")
        # Plots comparing features across clusters will be shown here
        visualize_cluster_analysis(feature_df.copy(), features_to_scale_and_cluster) # Pass copy

        # Step 9: Dimensionality Reduction
        print("\n=== Step 9: Dimensionality Reduction ===")
        feature_df = perform_dimensionality_reduction(feature_df, actual_scaled_cols_created)

        # Step 10: Visualize Clusters in 2D
        print("\n=== Step 10: Visualizing Clusters in 2D ===")
        # PCA and t-SNE plots will be shown here
        visualize_clusters_2d(feature_df.copy()) # Pass copy

        # Step 11: Save Final DataFrame
        print("\n=== Step 11: Saving Final Results ===")
        print(f"\n--- Saving final DataFrame to '{output_filename}' ---")
        try:
            # Define columns to include in the final output CSV
            output_columns = ['Customer ID'] + features_to_scale_and_cluster + ['Cluster', 'PCA1', 'PCA2', 'TSNE1', 'TSNE2']
            # Ensure only columns that actually exist are included (e.g., if PCA/TSNE failed)
            output_columns_exist = [col for col in output_columns if col in feature_df.columns]
            # Save the DataFrame
            feature_df.to_csv(output_filename, index=False, columns=output_columns_exist)
            print(f"  - Successfully saved {len(feature_df)} rows and {len(output_columns_exist)} columns.")
            print(f"  - Columns saved: {output_columns_exist}")
        except Exception as e:
            print(f"  - ERROR saving DataFrame to CSV: {e}")

    except Exception as e:
        # Catch any unexpected errors during the main pipeline execution
        print(f"\n!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!")
        print(f"!!!!!!!!!! An unhandled error occurred during main execution !!!!!!!!!")
        print(f"Error Type: {type(e).__name__}")
        print(f"Error Message: {e}")
        print("\nTraceback:")
        traceback.print_exc() # Print detailed traceback
        print(f"!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!")

    finally:
        # --- End of Execution ---
        # Final cleanup and timing
        gc.collect()
        end_time_main = datetime.datetime.now()
        print(f"\n===== Main Execution ended at: {end_time_main} =====")
        print(f"===== Total Duration: {end_time_main - start_time_main} =====")

# --- Script Entry Point ---
if __name__ == "__main__":
    # This block ensures the main function runs when the script is executed directly
    main()
```

**1. Elbow Method:**
The Elbow plot shows the Within-Cluster Sum of Squares (WCSS) for different values of `k`. We look for the "elbow" point where the rate of decrease in WCSS sharply slows down, suggesting diminishing returns from adding more clusters.

**2. Silhouette Score:**
The Silhouette score measures how similar an object is to its own cluster compared to other clusters. Scores range from -1 to 1, where a high value indicates that the object is well matched to its own cluster and poorly matched to neighboring clusters.
